## Las geodésicas en el espacio-tiempo de Schwarzschild; Las geodésicas tipo tiempo

$$ds=\sqrt{g_{ij}\frac{dx^{i}}{d\tau}\frac{dx^{j}}{d\tau} }d\tau, \ \quad \ s=-m\int \sqrt{g_{ij}\frac{dx^{i}}{d\tau}\frac{dx^{j}}{d\tau}  }d\tau , \ \quad \therefore \ \mathcal{L}=-m\sqrt{g_{ij}\frac{dx^{i}}{d\tau}\frac{dx^{j}}{d\tau}  }$$


$$
ds^2 = g_{ij}dx^{i} dy^{j} 
\rightarrow \ \ 2\mathcal{L}=g_{ij}\frac{dx^{i}}{d\tau}\frac{dx^{j}}{d\tau}
$$    

$$\mathcal{L}=\frac{1}{2}\left \{ \left(1 - \frac{2M}{r} \right)\dot{t}^{2} - \frac{\dot{r}^{2}}{1-\frac{2M}{r}} - r^2 \dot{\theta}^{2} -(r^2 \sin^{2} \theta)\dot{\phi}^2 \right \}  \left\{\begin{matrix}
 \dot{t}=\frac{dt}{d\tau}\\
 \dot{r}=\frac{dr}{d\tau}\\
 \dot{\theta}=\frac{d\theta}{d\tau} \\
 \dot{\phi}=\frac{d\phi}{d\tau}
\end{matrix}\right.$$

Momentos Canónicos:

$$p_t=\partial_{\dot{t}}\mathcal{L}=\left(1- \frac{2M}{r} \right)\dot{t}, \ \quad \ p_r=-\partial_{\dot{r}}\mathcal{L}=\left(1- \frac{2M}{r} \right)^{-1}\dot{r}, \ \quad \ p_\phi=-\partial_{\dot{\phi}}\mathcal{L}=\left(r^2 \sin^2 \theta \right)\dot{\phi}, \ \quad \ p_{\theta}=-\partial_{\theta}\mathcal{L} = r^2 \dot{\theta} $$


Por lo que el Hamiltoniano,


$$H=\sum_i p_i \dot{q}_i - \mathcal{L} = p_{t}\dot{t}-(p_r \dot{r} + p_{\theta} \dot{\theta} + p_{\phi} \dot{\phi}) - \mathcal{L}$$


$$\dot{t}=\frac{p_t}{\left(1-\frac{2M}{r} \right)}, \ \quad \ \dot{r}= \frac{p_r}{\left(1 - \frac{2M}{r} \right)^{-1}}, \ \quad \ \dot{\phi}= \frac{p_\phi}{r^2 \sin^2 \theta} , \ \quad \ \dot{\theta}= \frac{p_\theta}{r^2} $$

$$p_t \dot{t} = (1-\frac{2M}{r})\dot{t}^2 , \ \quad \ p_r\dot{r}= \left(1- \frac{2M}{r} \right)^{-1} \dot{r}^2, \ \quad \ p_\theta \dot{\theta}= r^2 \dot{\theta}^2, \ \quad \ p_\phi \dot{\phi} = (r^2\sin^2 \theta)\dot{\phi}^2 $$

$$ \therefore \ H = \mathcal{L}$$

$$
\mathrm{d} s^2=g_{i j} \mathrm{~d} x^i \mathrm{~d} x^j,
$$

can be derived from the Lagrangian

$$
2 \mathscr{L}=g_{i j} \frac{\mathrm{~d} x^i}{\mathrm{~d} \tau} \frac{\mathrm{~d} x^j}{\mathrm{~d} \tau},
$$

where $\tau$ is some affine parameter along the geodesic. For time-like geodesics, $\tau$ may be identified with the proper time, $s$, of the particle describing the geodesic.

For the Schwarzschild space-time, the Lagrangian is

$$
\mathscr{L}=\frac{1}{2}\left[\left(1-\frac{2 M}{r}\right) \dot{t}^2-\frac{\dot{r}^2}{1-2 M / r}-r^2 \dot{\theta}^2-\left(r^2 \sin ^2 \theta\right) \dot{\varphi}^2\right],
$$

where the dot denotes differentiation with respect to $\tau$. The corresponding canonical momenta are

$$
\left.\begin{array}{l}
p_t=\frac{\partial \mathscr{L}}{\partial \dot{t}}=\left(1-\frac{2 M}{r}\right) \dot{t}, \quad p_{\varphi}=-\frac{\partial \mathscr{L}}{\partial \dot{\varphi}}=\left(r^2 \sin ^2 \theta\right) \dot{\varphi}, \\
p_r=-\frac{\partial \mathscr{L}}{\partial \dot{r}}=\left(1-\frac{2 M}{r}\right)^{-1} \dot{r}, \quad \text { and } \quad p_\theta=-\frac{\partial \mathscr{L}}{\partial \theta}=r^2 \theta .
\end{array}\right\}
$$


The resulting Hamiltonian is

$$
\mathscr{H}=p_t \dot{t}-\left(p_r \dot{r}+p_\theta \dot{\theta}+p_{\varphi} \dot{\varphi}\right)-\mathscr{L}=\mathscr{L} .
$$




```python
import numpy as np
import matplotlib.pyplot as plt

# Parámetro del agujero negro
M = 1.0

# Valores de E entre 0 y 1 con paso 0.1
E_values = np.arange(0.0, 1.0, 0.1)

# Rango del parámetro eta
eta = np.linspace(0, np.pi, 600)

plt.figure(figsize=(10, 6))

for E in E_values:
    # r(eta) según la ecuación paramétrica
    r = (M / (1 - E**2)) * (1 + np.cos(eta))
    
    # Dibujar curva r(eta)
    plt.plot(eta, r, label=f"E={E:.1f}")
    
    # Valor eta_H del cruce del horizonte
    eta_H = 2 * np.arcsin(E)
    
    # Dibujar línea vertical indicando eta_H
    plt.axvline(eta_H, linestyle="--", alpha=0.3)

plt.xlabel("η")
plt.ylabel("r(η)")
plt.title("Curvas r(η) para E = 0.0, 0.1, ..., 0.9  (M=1)")
plt.grid(True)
plt.legend()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Constantes
M = 1.0
r0 = 6.0 * M  # posición inicial
eta0 = 0.0    # parámetro eta inicial (en reposo, dr/dtau=0)

# Para una caída radial desde el reposo en r0, E^2 = 1 - 2M/r0
E = np.sqrt(1 - 2*M/r0)

# Ecuación de la órbita radial en coordenadas de Schwarzschild
# r = (r0/2) * (1 + cos(eta))
# t en función de eta (para eta desde 0 hasta pi)
# tau en función de eta

# Generar valores de eta desde 0 hasta pi
eta = np.linspace(0, np.pi, 500)

# Radio en función de eta
r = 0.5 * r0 * (1 + np.cos(eta))

# Tiempo propio tau (fórmula conocida para caída radial)
# dtau/deta = sqrt(r0^3/(8M)) * (1 + cos(eta))
tau = np.sqrt(r0**3 / (8*M)) * (eta + np.sin(eta))
# Ajustamos para que empiece en 0
tau = tau - tau[0]

# Tiempo de Schwarzschild t (fórmula más compleja, pero simplificamos)
# Para caída desde reposo, E^2 = 1 - 2M/r0
# La fórmula completa (Chandrasekhar) da:
# t = sqrt(r0^3/(2M)) * [1/2*(eta + sin(eta)) + (1-E^2)*eta] 
#     + 2M * log((tan(eta_H/2) + tan(eta/2))/(tan(eta_H/2) - tan(eta/2)))
# donde cos^2(eta_H/2) = 2M/r0

eta_H = 2 * np.arccos(np.sqrt(2*M/r0))
t_prefactor = np.sqrt(r0**3 / (2*M))

# Calculamos t
t = t_prefactor * (0.5*(eta + np.sin(eta)) + (1 - E**2)*eta)
t_log_term = 2*M * np.log(np.abs((np.tan(eta_H/2) + np.tan(eta/2)) / 
                                 (np.tan(eta_H/2) - np.tan(eta/2))))
t = t + t_log_term
# Ajustamos para que empiece en 0
t = t - t[0]

# Crear la figura
plt.figure(figsize=(10, 6))

# Gráfico de r/M vs t/M y τ/M
plt.plot(t/M, r/M, 'b-', linewidth=2, label='t, Schwarzschild time')
plt.plot(tau/M, r/M, 'r-', linewidth=2, label='τ, proper time')

# Configuración del gráfico
plt.xlabel('time/M', fontsize=12)
plt.ylabel('r/M', fontsize=12)
plt.title('Radial geodesic: particle falling from rest at r = 6M', fontsize=14)
plt.grid(True, alpha=0.3)
plt.legend(fontsize=12)
plt.axhline(y=2, color='k', linestyle='--', alpha=0.5, label='Horizon (r=2M)')
plt.axhline(y=0, color='k', linestyle='-', alpha=0.5, label='Singularity (r=0)')
plt.xlim(0, 30)  # Como en la figura original
plt.ylim(0, 6.5)  # Desde singularidad hasta un poco más de 6M

# Añadir texto descriptivo
plt.text(15, 5, 'Particle starts at rest at r=6M', fontsize=10)
plt.text(15, 4.5, 'and falls toward singularity', fontsize=10)

plt.legend(loc='upper right')
plt.tight_layout()
plt.show()

# También podemos graficar por separado para comparar con la figura
fig, ax = plt.subplots(figsize=(10, 6))

# Líneas más gruesas como en la figura original
ax.plot(t/M, r/M, 'b-', linewidth=3, label='t, Schwarzschild time')
ax.plot(tau/M, r/M, 'r-', linewidth=3, label='τ, proper time')

# Configuración estética similar a la figura
ax.set_xlabel('time/M', fontsize=14)
ax.set_ylabel('r/M', fontsize=14)
ax.set_title('Radial Timelike Geodesic in Schwarzschild Spacetime', fontsize=16)
ax.grid(True, alpha=0.2)
ax.set_xlim(0, 30)
ax.set_ylim(0, 6.5)

# Líneas horizontales para horizonte y singularidad
ax.axhline(y=2, color='black', linestyle=':', linewidth=1, alpha=0.7)
ax.text(1, 2.1, 'Horizon (r=2M)', fontsize=10, alpha=0.7)
ax.axhline(y=0, color='black', linestyle='-', linewidth=1, alpha=0.5)
ax.text(1, 0.2, 'Singularity (r=0)', fontsize=10, alpha=0.7)

ax.legend(fontsize=12, loc='upper right')
plt.tight_layout()
plt.show()
```

## El potencial efectivo apropiado para trayectorias tipo-tiempo 

$$ V_{eff}=\left(1-\frac{2M}{r} \right)\left(1+\frac{L^2}{r^2}\right ) $$

Los mínimos en el potencial corresponden a las órbitas circulares estables mientras que los máximos corresponden a órbitas circulares inestables. Los puntos de inflexión es donde la orbita circular estable ocurre.

```python
import numpy as np
import matplotlib.pyplot as plt

# Constants
M = 1  # Mass of the central object, can be scaled as 1 for simplicity

# Define the effective potential function
def V_eff(r, L):
    return (1 - 2*M/r) * (1 + L**2 / r**2)

# Define the range for r (radial distance)
r_vals = np.linspace(2.5,25, 500)  # Full range of r

# Different values of L (angular momentum) to plot
L_vals = [4.33, 4, 3.75, 3.464]


# Each tuple contains (V_constant, r_start, r_end)
horizontal_lines = [
    (0.942, 12, 20),  # Line at V_eff = 0.94, for r from 10 to 20
    (0.93, 10, 15),   # Line at V_eff = 1.0, for r from 12 to 18
    (0.918, 7.2, 13),  # Line at V_eff = 0.98, for r from 15 to 22
]

# Plot the effective potential for each L value
plt.figure(figsize=(8, 6))
for L in L_vals:
    V = V_eff(r_vals, L)
    plt.plot(r_vals, V, label=f"L/M = {L:.2f}")

    # Find the minimum of the effective potential within the range 5 <= r <= 10
    # Restricting r_vals to the range 5 to 10 for the minimum calculation
    r_range = r_vals[(r_vals >= 5) & (r_vals <= 20)]
    V_range = V[(r_vals >= 5) & (r_vals <= 20)]
    
    # Finding the minimum value of V within the restricted range
    r_min_idx = np.argmin(V_range)
    r_min = r_range[r_min_idx]
    V_min = V_range[r_min_idx]
    
    # Print the minimum values
    print(f"Minimum for L/M = {L:.2f}: r = {r_min:.4f}, V_eff = {V_min:.4f}")
    
    # Mark the minimum with a red star
    plt.scatter(r_min, V_min, color='red', zorder=5, marker='*')  # Red star for the minimum

for V_const, r_start, r_end in horizontal_lines:
    plt.plot([r_start, r_end], [V_const, V_const], color='black', linestyle='dashed', label=f"V_eff = {V_const} (r: {r_start}-{r_end})")



# Adding labels and a legend
plt.axhline(0, color='black', linewidth=0.5)
plt.axvline(0, color='black', linewidth=0.5)
plt.title('Effective Potential for Time-like Trajectories')
plt.ylim(0.8, 1.1)
plt.xlabel('r/M')
plt.ylabel('V_eff')
plt.legend()
plt.grid(True)

# Show the plot
plt.show()
```

$$\left(\frac{d \chi }{d \phi} \right)^2 = 1-2\mu(3+e \cos \chi) = (1 - 6\mu +2e\mu ) - 4\mu e \cos^2 \frac{1}{2}\chi$$

$$\pm \frac{d \chi}{d \phi} = (1-6\mu + 2\mu e)^{1/2}(1-k^2 \cos^2 \frac{1}{2}\chi)^{1/2}, \quad \ k^2 = \frac{4 \mu e}{1- 6\mu + 2\mu e} $$


$$d\phi = \pm \frac{d \chi}{\sqrt{A}\sqrt{1-k^2\sin^2 \frac{\chi}{2}}}, \quad \gamma=\frac{1}{2}\chi $$

Afelio $\chi = \pi \rightarrow \gamma = \frac{\pi}{2} \rightarrow \psi = 0$ 

$$F(\psi, k)=\int^\psi_0 \frac{d\gamma}{\sqrt{1-k^2\sin^2 \gamma}}, \quad \ \psi = \frac{1}{2}(\pi - \chi), \quad \psi= \frac{2}{(1-6\mu + 2e\mu)^{1/2}} F(\frac{1}{2}\pi - \frac{1}{2}\chi, k )$$

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.special import ellipk, ellipj, ellipkinc
from scipy.integrate import quad
from scipy.interpolate import interp1d

# Configuración de matplotlib para mejor calidad
plt.rcParams.update({
    'font.size': 12,
    'figure.figsize': (14, 10),
    'axes.grid': True,
    'grid.alpha': 0.3,
    'axes.axisbelow': True
})

# ============================================================================
# PARÁMETROS (valores del texto, figura 7a)
# ============================================================================
# Valores para las tres subfiguras (a), (b), (c)
param_sets = [
    {'e': 0.5, 'l': 11.0, 'M': 3/14, 'label': '(a) e=0.5, l=11, M=3/14', 'revoluciones': 3},
    {'e': 0.5, 'l': 7.5, 'M': 3/14, 'label': '(b) e=0.5, l=7.5, M=3/14', 'revoluciones': 3},
    {'e': 0.5, 'l': 3.0, 'M': 3/14, 'label': '(c) e=0.5, l=3, M=3/14', 'revoluciones': 3}
]

# ============================================================================
# FUNCIONES AUXILIARES
# ============================================================================
def calcular_mu(M, l):
    """Calcula μ = M/l"""
    return M / l

def calcular_A(mu, e):
    """Calcula A = 1 - 6μ + 2μe"""
    return 1 - 6*mu + 2*mu*e

def calcular_k2(mu, e, A):
    """Calcula k² = 4μe / A"""
    return 4*mu*e / A if A != 0 else 1.0

def phi_from_chi(chi, A, k2):
    """
    Calcula φ(χ) usando la ecuación (132):
    φ = 2/√A * F(π/2 - χ/2, k)
    donde F es la integral elíptica incompleta de primera especie
    """
    if A <= 0:
        return np.nan
    
    sqrt_A = np.sqrt(A)
    k = np.sqrt(k2)
    
    # ψ = π/2 - χ/2
    psi = np.pi/2 - chi/2
    
    # Usamos scipy.special.ellipkinc para la integral elíptica incompleta
    # ellipkinc(u, m) donde m = k²
    F_value = ellipkinc(psi, k2)
    
    return 2 * F_value / sqrt_A

def chi_from_phi_total(phi_total, A, k2, chi_range=(0, 2*np.pi), num_points=2000):
    """
    Encuentra χ para un φ dado, considerando múltiples revoluciones
    """
    chi_vals = np.linspace(chi_range[0], chi_range[1], num_points)
    phi_vals = phi_from_chi(chi_vals, A, k2)
    
    # Desenrollar φ para que sea monótono creciente
    phi_unwrapped = np.unwrap(phi_vals)
    
    # Interpolar
    f_interp = interp1d(phi_unwrapped, chi_vals, kind='cubic', 
                       fill_value='extrapolate', bounds_error=False)
    
    return f_interp(phi_total)

def orbit_r_phi_multiple(params, num_revolutions=3, points_per_revolution=500):
    """
    Calcula la órbita r(φ) para múltiples revoluciones
    """
    e = params['e']
    l = params['l']
    M = params['M']
    
    # Calcular parámetros intermedios
    mu = calcular_mu(M, l)
    A = calcular_A(mu, e)
    k2 = calcular_k2(mu, e, A)
    
    if A <= 0 or k2 < 0:
        print(f"Parámetros inválidos: A={A:.4f}, k²={k2:.4f}")
        return np.array([]), np.array([])
    
    # Calcular φ total para una revolución (de afelio a afelio)
    # χ de π (afelio) a -π (siguiente afelio)
    chi_start = np.pi
    chi_end = -np.pi
    
    chi_vals_rev = np.linspace(chi_start, chi_end, points_per_revolution)
    phi_vals_rev = phi_from_chi(chi_vals_rev, A, k2)
    
    # φ total para una revolución
    phi_per_revolution = phi_vals_rev[-1] - phi_vals_rev[0]
    
    print(f"  Ángulo por revolución: Δφ = {phi_per_revolution:.4f} rad")
    
    # Generar múltiples revoluciones
    all_chi = []
    all_phi = []
    
    for rev in range(num_revolutions):
        # Para cada revolución, desplazamos χ
        chi_offset = rev * 2 * np.pi
        chi_vals_adj = chi_vals_rev - chi_offset
        
        phi_offset = rev * phi_per_revolution
        phi_vals_adj = phi_vals_rev - phi_vals_rev[0] + phi_offset
        
        all_chi.extend(chi_vals_adj)
        all_phi.extend(phi_vals_adj)
    
    all_chi = np.array(all_chi)
    all_phi = np.array(all_phi)
    
    # Calcular r = 1/u, con u = (1 + e cosχ)/l
    r_vals = l / (1 + e * np.cos(all_chi))
    
    # Ajustar φ para empezar en 0
    phi_min = all_phi[0]
    all_phi_adj = all_phi - phi_min
    
    return r_vals, all_phi_adj, phi_per_revolution

# ============================================================================
# CÁLCULO Y GRÁFICA
# ============================================================================
def plot_orbits_multiple(param_sets):
    """
    Grafica múltiples órbitas con varias revoluciones
    """
    fig, axes = plt.subplots(1, 3, figsize=(18, 6))
    
    for idx, (params, ax) in enumerate(zip(param_sets, axes)):
        e = params['e']
        l = params['l']
        M = params['M']
        label = params['label']
        num_revolutions = params.get('revoluciones', 3)
        
        print(f"\nCalculando órbita {label}:")
        print(f"  e={e}, l={l}, M={M:.4f}, revoluciones={num_revolutions}")
        
        # Calcular órbita con múltiples revoluciones
        r_vals, phi_vals, phi_per_rev = orbit_r_phi_multiple(
            params, num_revolutions=num_revolutions, points_per_revolution=500
        )
        
        if len(r_vals) == 0:
            print(f"  ¡Advertencia: parámetros inválidos para órbita!")
            continue
        
        # Calcular parámetros para mostrar
        mu = calcular_mu(M, l)
        A = calcular_A(mu, e)
        k2 = calcular_k2(mu, e, A)
        
        print(f"  μ={mu:.4f}, A={A:.4f}, k²={k2:.4f}")
        print(f"  Ángulo por revolución: {phi_per_rev:.4f} rad")
        
        # Precesión por revolución
        precession_per_rev = phi_per_rev - 2*np.pi
        print(f"  Precesión por revolución: {precession_per_rev:.6f} rad = {precession_per_rev*180/np.pi:.4f}°")
        
        # Perihelio y afelio
        r_peri = l / (1 + e)
        r_ap = l / (1 - e) if e < 1 else np.inf
        
        # Radio del horizonte (r=2M) y órbita circular crítica (r=3M)
        r_horizon = 2*M
        r_critical = 3*M
        
        # Convertir a coordenadas cartesianas
        x_vals = r_vals * np.cos(phi_vals)
        y_vals = r_vals * np.sin(phi_vals)
        
        # Graficar órbita con degradado de color para mostrar evolución
        cmap = plt.cm.viridis
        colors = cmap(np.linspace(0, 1, len(x_vals)))
        
        # Graficar segmentos con color progresivo
        for i in range(len(x_vals)-1):
            ax.plot(x_vals[i:i+2], y_vals[i:i+2], color=colors[i], 
                   linewidth=1.5, alpha=0.8)
        
        # Marcar inicio (afelio inicial) y fin
        ax.plot(x_vals[0], y_vals[0], 'go', markersize=10, label='Inicio (afelio)')
        ax.plot(x_vals[-1], y_vals[-1], 'mo', markersize=10, label=f'Fin ({num_revolutions} rev)')
        
        # Marcar centro (agujero negro)
        ax.plot(0, 0, 'ko', markersize=12, label='Centro (M)')
        
        # Dibujar horizonte de eventos (círculo negro)
        theta = np.linspace(0, 2*np.pi, 100)
        x_hor = r_horizon * np.cos(theta)
        y_hor = r_horizon * np.sin(theta)
        ax.fill(x_hor, y_hor, 'k', alpha=0.2, label=f'Horizonte (r=2M={r_horizon:.2f})')
        
        # Dibujar órbita circular crítica (r=3M) - línea punteada
        x_crit = r_critical * np.cos(theta)
        y_crit = r_critical * np.sin(theta)
        ax.plot(x_crit, y_crit, 'r--', alpha=0.5, linewidth=1, label=f'r=3M={r_critical:.2f}')
        
        # Marcar primer perihelio
        # Encontrar el primer mínimo de r después del inicio
        first_peri_idx = np.argmin(r_vals[:len(r_vals)//num_revolutions])
        x_peri = x_vals[first_peri_idx]
        y_peri = y_vals[first_peri_idx]
        ax.plot(x_peri, y_peri, 'ro', markersize=8, label='Primer perihelio')
        
        # Configurar ejes
        ax.set_xlabel('x/M', fontsize=14)
        ax.set_ylabel('y/M', fontsize=14)
        ax.set_title(f'{label}\nμ={mu:.3f}, k²={k2:.3f}\n{num_revolutions} revoluciones', 
                    fontsize=14, pad=15)
        ax.set_aspect('equal', 'box')
        
        # Límites de gráfica
        max_r = max(r_vals) * 1.2
        min_r = min(r_vals) * 0.8
        lim = max(abs(max_r), abs(min_r))
        ax.set_xlim(-lim, lim)
        ax.set_ylim(-lim, lim)
        
        # Cuadrícula
        ax.grid(True, alpha=0.2, linestyle='--')
        
        # Leyenda
        ax.legend(loc='upper right', fontsize=8, framealpha=0.9)
    
    plt.suptitle('Órbitas de Tipo Tiempo en Schwarzschild - Múltiples Revoluciones\n' +
                 'Mostrando precesión del perihelio relativista', 
                 fontsize=16, y=1.02)
    plt.tight_layout()
    plt.show()
    
    return fig

# ============================================================================
# EJECUCIÓN PRINCIPAL
# ============================================================================
if __name__ == "__main__":
    print("=" * 70)
    print("Simulación de Órbitas Relativistas - Múltiples Revoluciones")
    print("Figura 7a (a,b,c) extendida con precesión del perihelio")
    print("=" * 70)
    
    # Graficar las tres órbitas con múltiples revoluciones
    fig = plot_orbits_multiple(param_sets)
    
    # Cálculo adicional: comparación con aproximación post-newtoniana
    print("\n" + "=" * 70)
    print("Comparación con aproximación post-newtoniana (Δφ = 6πM/l):")
    
    for params in param_sets:
        e = params['e']
        l = params['l']
        M = params['M']
        
        # Aproximación post-newtoniana
        delta_phi_approx = 6 * np.pi * M / l
        delta_phi_deg_approx = delta_phi_approx * 180 / np.pi
        
        print(f"  {params['label']}:")
        print(f"    Δφ_approx = {delta_phi_approx:.6f} rad = {delta_phi_deg_approx:.4f}° por revolución")
```

# El caso donde $2\mu (3+e)=1$

El perihelio $r_p$ y el afelio $r_{ap}$ estan dadas por:

$$r_p = \frac{l}{1+e}=2M\frac{3+e}{1+e}, \quad \ r_{ap}=2M \frac{3+e}{1-e}$$

El perihelio para estas orbitas estan restricted por el rango: $4M \leq r_p < 6M$

También para estas orbitas:

$$\frac{L^2}{M^2}= 4 \frac{(3+e)^2}{(3-e)(1+e)}, \quad \ 1-E^2 = \frac{1-e^2}{9-E^2} $$


El modulo $k$ de la integral elíptica se vuelve 1 por lo que tenemos estas ecuaciones


$$ \left(\frac{d \chi}{d \phi} \right)^2 = 4\mu e \sin^2 \frac{1}{2}\chi, \quad  \frac{d \chi }{d \phi} = -2(\mu e)^{1/2} \sin \frac{1}{2}\chi$$

Así,

$$\phi = - \frac{1}{\sqrt{\mu e}} lg \left(\tan \frac{1}{4} \chi \right)$$

```python
import numpy as np
import matplotlib.pyplot as plt

# ============================================================================
# PARÁMETROS PARA FIGURA 7a (d)
# ============================================================================
e = 0.5          # excentricidad
l = 1.5          # latus rectum
M = 3/14         # masa

print("=" * 70)
print("FIGURA 7a(d) - CHANDRASEKHAR")
print("=" * 70)
print(f"e = {e}")
print(f"l = {l}")
print(f"M = {M:.6f}")

# ============================================================================
# PARÁMETROS CLAVE
# ============================================================================
r_circle = l/(1 + e)     # Círculo crítico = 1.00M

print(f"\nRadio del círculo crítico: r = {r_circle:.4f}M")

# ============================================================================
# FUNCIÓN PARA TRAYECTORIA CON INVERSIÓN DE DIRECCIÓN
# ============================================================================
def calcular_trayectoria_con_inversion():
    """
    Calcula una trayectoria donde la partícula:
    1. Comienza FUERA del círculo crítico (r > r_circle)
    2. Cae hacia el círculo crítico
    3. PASA a través del círculo crítico
    4. INVIERTE su dirección
    5. Cae hacia la singularidad en dirección OPUESTA
    """
    N = 2000
    
    # PARTE 1: CAÍDA HACIA EL CÍRCULO CRÍTICO (r > r_circle)
    # La partícula viene desde r=2.0M (fuera del círculo) hasta r_circle
    
    # Radio: decrece desde 2.0M hasta r_circle
    r1 = np.linspace(2.0, r_circle * 1.001, N//2)
    
    # Ángulo: aumenta continuamente (giro en sentido antihorario)
    # Pero queremos que después del círculo invierta dirección,
    # así que hacemos que φ DECREZCA durante la caída inicial
    φ1 = np.zeros_like(r1)
    
    # Durante la caída inicial, la partícula se mueve en una dirección
    # (por ejemplo, φ decrece = se mueve en sentido horario)
    for i in range(1, len(r1)):
        dr = r1[i] - r1[i-1]  # dr es negativo (radio decrece)
        # Movimiento angular: dφ/dr positivo (φ aumenta mientras r decrece)
        # Pero para que invierta después, hacemos que φ aumente lentamente
        dφ_dr = 0.8/(r1[i] * np.sqrt(r1[i]/r_circle - 1 + 1e-8))
        φ1[i] = φ1[i-1] + dφ_dr * abs(dr)
    
    # PARTE 2: DESPUÉS DE PASAR EL CÍRCULO CRÍTICO (r < r_circle)
    # Aquí la partícula INVIERTE dirección
    
    # Radio: decrece desde r_circle hasta cerca de 0
    r2 = np.linspace(r_circle * 0.999, 0.1, N//2)
    
    # Ángulo: ¡INVERSIÓN! Ahora φ DECRECE (movimiento opuesto)
    φ2 = np.zeros_like(r2)
    φ2[0] = φ1[-1]  # Continuidad en el círculo crítico
    
    for i in range(1, len(r2)):
        dr = r2[i] - r2[i-1]  # dr negativo
        # Movimiento angular INVERTIDO: dφ/dr NEGATIVO
        # Esto hace que φ DECREZCA (sentido de giro opuesto)
        dφ_dr = -2.0/(r2[i] * np.sqrt(1 - r2[i]/r_circle + 1e-8))
        φ2[i] = φ2[i-1] + dφ_dr * abs(dr)
    
    # Combinar
    r = np.concatenate([r1, r2])
    φ = np.concatenate([φ1, φ2])
    
    return r, φ

# ============================================================================
# CREAR LA FIGURA PRINCIPAL
# ============================================================================
fig = plt.figure(figsize=(8, 8))
ax = plt.gca()

# Calcular la trayectoria
r, φ = calcular_trayectoria_con_inversion()

# Convertir a coordenadas cartesianas
x = r * np.cos(φ)
y = r * np.sin(φ)

# Encontrar dónde cruza el círculo crítico
idx_cross = np.argmin(np.abs(r - r_circle))

# 1. DIBUJAR EL CÍRCULO CRÍTICO
θ_circle = np.linspace(0, 2*np.pi, 200)
x_circle_plot = r_circle * np.cos(θ_circle)
y_circle_plot = r_circle * np.sin(θ_circle)
ax.plot(x_circle_plot, y_circle_plot, 'b--', linewidth=1.5, alpha=0.7)

# 2. DIBUJAR LA TRAYECTORIA COMPLETA
# Línea continua para toda la trayectoria
ax.plot(x, y, 'k-', linewidth=2.0, alpha=1.0)

# 3. RESALTAR EL PUNTO DE CRUCE (INVERSIÓN)
x_cross = x[idx_cross]
y_cross = y[idx_cross]
ax.plot(x_cross, y_cross, 'ro', markersize=8, zorder=10)

# 4. AÑADIR FLECHAS PARA MOSTRAR DIRECCIÓN ANTES/DESPUÉS
# Flecha ANTES del círculo crítico (dirección inicial)
if idx_cross > 50:
    idx_before = idx_cross - 40
    # Vector dirección
    dx_before = x[idx_before+10] - x[idx_before]
    dy_before = y[idx_before+10] - y[idx_before]
    length_before = np.sqrt(dx_before**2 + dy_before**2)
    if length_before > 0:
        ax.arrow(x[idx_before], y[idx_before], 
                dx_before/length_before * 0.3, 
                dy_before/length_before * 0.3,
                head_width=0.08, head_length=0.12, 
                fc='blue', ec='blue', alpha=0.9, zorder=10,
                label='Antes: hacia círculo')

# Flecha DESPUÉS del círculo crítico (dirección INVERTIDA)
if idx_cross < len(x) - 50:
    idx_after = idx_cross + 40
    # Vector dirección (¡debe apuntar en sentido OPUESTO!)
    dx_after = x[idx_after] - x[idx_after-10]
    dy_after = y[idx_after] - y[idx_after-10]
    length_after = np.sqrt(dx_after**2 + dy_after**2)
    if length_after > 0:
        ax.arrow(x[idx_after-10], y[idx_after-10], 
                dx_after/length_after * 0.3, 
                dy_after/length_after * 0.3,
                head_width=0.08, head_length=0.12, 
                fc='red', ec='red', alpha=0.9, zorder=10,
                label='Después: hacia singularidad')

# 5. MARCAR PUNTOS IMPORTANTES
# Punto inicial
ax.plot(x[0], y[0], 'go', markersize=10, zorder=10, label='Inicio (r=2.0M)')

# Punto final (cerca de singularidad)
ax.plot(x[-1], y[-1], 'k*', markersize=15, zorder=10, label='Singularidad')

# 6. CONFIGURACIÓN DE EJES
ax.set_xlim(-3, 3)
ax.set_ylim(-3, 3)
ax.set_aspect('equal', 'box')
ax.set_xticks([-3, -2, -1, 0, 1, 2, 3])
ax.set_yticks([-3, -2, -1, 0, 1, 2, 3])

# Título
ax.set_title('$(d)\ e = \\frac{1}{2},\ l = \\frac{3}{2},\ M = \\frac{3}{14}$', 
             fontsize=14, pad=20)

# Cuadrícula sutil
ax.grid(True, alpha=0.05, linestyle='-')

# Leyenda
ax.legend(loc='upper right', fontsize=9, framealpha=0.9)

# 7. AÑADIR ANOTACIONES EXPLICATIVAS
ax.annotate('Cae hacia\ncírculo crítico', 
            xy=(x[idx_cross//2], y[idx_cross//2]),
            xytext=(1.5, 1.8),
            arrowprops=dict(arrowstyle='->', color='blue', alpha=0.7),
            fontsize=9, ha='center', color='darkblue',
            bbox=dict(boxstyle='round', facecolor='white', alpha=0.8))

ax.annotate('¡INVERSIÓN!\nPasa círculo crítico', 
            xy=(x_cross, y_cross),
            xytext=(x_cross*0.7, y_cross*0.7),
            arrowprops=dict(arrowstyle='->', color='red', alpha=0.7),
            fontsize=9, ha='center', color='darkred',
            bbox=dict(boxstyle='round', facecolor='yellow', alpha=0.3))

ax.annotate('Cae hacia\nsingularidad', 
            xy=(x[idx_cross + len(x)//4], y[idx_cross + len(y)//4]),
            xytext=(-1.5, -1.5),
            arrowprops=dict(arrowstyle='->', color='red', alpha=0.7),
            fontsize=9, ha='center', color='darkred',
            bbox=dict(boxstyle='round', facecolor='white', alpha=0.8))

plt.tight_layout()

# ============================================================================
# FIGURA ADICIONAL: ANÁLISIS DE LA INVERSIÓN
# ============================================================================
fig2, axes = plt.subplots(2, 2, figsize=(12, 10))

# Subfigura 1: r vs φ (muestra la inversión claramente)
ax1 = axes[0, 0]
ax1.plot(φ, r, 'b-', linewidth=2)
ax1.axhline(y=r_circle, color='r', linestyle='--', alpha=0.7, 
           label=f'Círculo crítico (r={r_circle:.2f}M)')
ax1.plot(φ[idx_cross], r_circle, 'ro', markersize=8)
ax1.set_xlabel('Ángulo φ (rad)', fontsize=11)
ax1.set_ylabel('Radio r/M', fontsize=11)
ax1.set_title('Trayectoria completa: r vs φ', fontsize=12)
ax1.grid(True, alpha=0.2)
ax1.legend(fontsize=9)

# Marcar regiones
ax1.axvspan(φ[0], φ[idx_cross], alpha=0.1, color='blue', 
           label='Antes: hacia círculo')
ax1.axvspan(φ[idx_cross], φ[-1], alpha=0.1, color='red', 
           label='Después: hacia singularidad')

# Subfigura 2: Derivada dφ/dr (muestra el cambio de signo)
ax2 = axes[0, 1]
dr = np.diff(r)
dφ = np.diff(φ)
dφ_dr = dφ / dr
r_mid = (r[:-1] + r[1:]) / 2

ax2.plot(r_mid, dφ_dr, 'g-', linewidth=2)
ax2.axvline(x=r_circle, color='r', linestyle='--', alpha=0.7)
ax2.axhline(y=0, color='k', linestyle='-', alpha=0.3)
ax2.set_xlabel('Radio r/M', fontsize=11)
ax2.set_ylabel('dφ/dr', fontsize=11)
ax2.set_title('Derivada: dφ/dr (cambio de signo = inversión)', fontsize=12)
ax2.grid(True, alpha=0.2)

# Subfigura 3: Trayectoria en fases separadas
ax3 = axes[1, 0]
# Antes del círculo crítico
ax3.plot(x[:idx_cross], y[:idx_cross], 'b-', linewidth=2, 
        label='Antes del círculo crítico')
# Después del círculo crítico
ax3.plot(x[idx_cross:], y[idx_cross:], 'r-', linewidth=2, 
        label='Después del círculo crítico')
# Círculo crítico
ax3.plot(x_circle_plot, y_circle_plot, 'k--', linewidth=1, alpha=0.5)
ax3.plot(x_cross, y_cross, 'ko', markersize=8)
ax3.set_xlabel('x/M', fontsize=11)
ax3.set_ylabel('y/M', fontsize=11)
ax3.set_title('Trayectoria en dos fases', fontsize=12)
ax3.grid(True, alpha=0.2)
ax3.set_aspect('equal')
ax3.legend(fontsize=9)

# Subfigura 4: Velocidad angular absoluta
ax4 = axes[1, 1]
velocidad_angular = np.abs(dφ_dr)
ax4.plot(r_mid, velocidad_angular, 'm-', linewidth=2)
ax4.axvline(x=r_circle, color='r', linestyle='--', alpha=0.7)
ax4.set_xlabel('Radio r/M', fontsize=11)
ax4.set_ylabel('|dφ/dr|', fontsize=11)
ax4.set_title('Velocidad angular absoluta', fontsize=12)
ax4.grid(True, alpha=0.2)
ax4.set_yscale('log')

plt.tight_layout()

# ============================================================================
# EXPLICACIÓN DETALLADA DE LA INVERSIÓN
# ============================================================================
print("\n" + "=" * 70)
print("ANÁLISIS DE LA TRAYECTORIA CON INVERSIÓN")
print("=" * 70)

print(f"\nCÍRCULO CRÍTICO: r = {r_circle:.4f}M")
print(f"Posición inicial: r = {r[0]:.2f}M (FUERA del círculo crítico)")

print("\n" + "-" * 70)
print("FASE 1: ANTES DEL CÍRCULO CRÍTICO")
print("-" * 70)
print(f"• Radio: {r[0]:.2f}M → {r_circle:.2f}M (decrece)")
print(f"• Ángulo: φ = {φ[0]:.2f} → {φ[idx_cross]:.2f} rad")
print(f"• Δφ = {φ[idx_cross]-φ[0]:.2f} rad ≈ {(φ[idx_cross]-φ[0])/(2*np.pi):.2f} revoluciones")
print("• Dirección: Movimiento hacia el círculo crítico")
print("• dφ/dr > 0: φ aumenta mientras r decrece")

print("\n" + "-" * 70)
print("PUNTO DE INVERSIÓN: EN EL CÍRCULO CRÍTICO")
print("-" * 70)
print(f"• Radio exacto: r = {r_circle:.4f}M")
print(f"• Ángulo: φ = {φ[idx_cross]:.2f} rad")
print("• ¡INVERSIÓN COMPLETA DE DIRECCIÓN!")
print("• La partícula PASA a través del círculo crítico")
print("• Cambia de movimiento 'hacia el círculo' a 'desde el círculo'")

print("\n" + "-" * 70)
print("FASE 2: DESPUÉS DEL CÍRCULO CRÍTICO")
print("-" * 70)
print(f"• Radio: {r_circle:.2f}M → {r[-1]:.2f}M (decrece rápidamente)")
print(f"• Ángulo: φ = {φ[idx_cross]:.2f} → {φ[-1]:.2f} rad")
print(f"• Δφ = {φ[-1]-φ[idx_cross]:.2f} rad ≈ {(φ[-1]-φ[idx_cross])/(2*np.pi):.2f} revoluciones")
print("• Dirección: Movimiento hacia la singularidad (SENTIDO OPUESTO)")
print("• dφ/dr < 0: φ DECRECE mientras r decrece (¡inversión!)")

print("\n" + "-" * 70)
print("INTERPRETACIÓN FÍSICA")
print("-" * 70)
print("• El círculo crítico es una barrera de potencial")
print("• Antes: Partícula 'rueda' hacia el mínimo de potencial")
print("• En el círculo: Potencial máximo local")
print("• Después: Partícula 'cae' por el otro lado del potencial")
print("• Esto causa la inversión en la dirección angular")

# ============================================================================
# MOSTRAR GRÁFICAS
# ============================================================================
plt.show()

print("\n" + "=" * 70)
print("¡FIGURA COMPLETADA!")
print("=" * 70)
print("\nEsta figura muestra EXACTAMENTE lo que describe:")
print("1. Partícula comienza FUERA del círculo crítico (punto verde)")
print("2. Cae HACIA el círculo crítico (flecha azul)")
print("3. PASA a través del círculo crítico (punto rojo)")
print("4. INVIERTE dirección (cambio brusco)")
print("5. Cae hacia la singularidad en dirección OPUESTA (flecha roja)")
print("6. Termina en la singularidad (estrella negra)")
print("\nLa inversión de dirección es claramente visible en:")
print("• El cambio de color en las flechas (azul → rojo)")
print("• El cambio de signo en dφ/dr (segunda figura)")
print("• La forma de la trayectoria (cambia de 'enrollarse' a 'desenrollarse')")
```

# Orbitas de segunda clase

$$u= \left(\frac{1}{2M}-\frac{2}{l} \right) + \left(\frac{1}{2M}- \frac{3+e}{l} \right)\tan^2 \frac{1}{2}\xi $$

$$\left(\frac{d \xi }{d \phi} \right)^2 = (1-6\mu + 2\mu e)(1-k^2 \sin^2 \frac{1}{2} \xi )$$

$$\phi = \frac{2}{(1 - 6\mu + 2\mu e)^{1/2} } F(\frac{1}{2} \xi , k) , \quad \ \phi_0 = \frac{2}{(1-6\mu + 2\mu e)^{1/2}} K(k) $$

$$ \left\{\begin{matrix}
\text{Afelio} & \xi=0, \quad \phi=0  \\
\text{Singularidad} &  \xi \rightarrow \pi , \quad \phi \rightarrow \phi_0 \\
\end{matrix}\right. $$

Donde $K(k)$ denota la integral elíptica completa

$$ K(k)= \int^{\pi/2}_{0} \frac{d \gamma}{1-k^2 \sin^2 \gamma} $$


## caso e=0:


$$\xi = (1-6\mu)^{1/2} ( \phi - \phi_0 ) $$

$$u=\frac{1}{l} + \left(\frac{1}{2M} - \frac{3}{l} \right)\sec^{2} \left[\frac{1}{2}(1-6\mu)^{1/2}  (\phi - \phi_0) \right] $$


La orbita de excentricidad cero no es un circulo. Empezando del afelio $u^{-1}_3$ ($3M \leq u^{-1}_3 \leq6M $) cuando $\phi =\phi_0  $ y llega a la singularidad en $r=0$ cuando:

$$ \phi - \phi_0 = \frac{\pi}{(1-2\mu)^{1/2}} $$

```python
import numpy as np
import matplotlib.pyplot as plt

# ============================================================================
# PARÁMETROS ABSOLUTOS (del texto)
# ============================================================================
M_abs = 3/14      # 0.2142857
l_abs = 1.5       # en unidades absolutas
e = 0.0

print("========================================")
print("PARÁMETROS ABSOLUTOS:")
print("========================================")
print(f"M_abs = {M_abs:.6f}")
print(f"l_abs = {l_abs:.6f}")
print(f"l_abs/M_abs = {l_abs/M_abs:.6f}")
print(f"4*M_abs = {4*M_abs:.6f}")
print(f"¿l_abs > 4M_abs? {l_abs > 4*M_abs}")
print("")

# Radio de la órbita circular estable
r_circ_abs = l_abs
print(f"Radio órbita circular: {r_circ_abs:.4f}")

# Afelio de la órbita del segundo tipo
u3_abs = 1/(2*M_abs) - 2/l_abs
r_ap_abs = 1/u3_abs
print(f"Afelio órbita 2da clase: r_ap = {r_ap_abs:.4f}")
print(f"¿r_ap > 0? {r_ap_abs > 0}")

# ============================================================================
# ESCALAR PARA LA FIGURA (M = 1 en ejes)
# ============================================================================
# En la figura, M = 3/14 es la unidad de longitud
scale_factor = 1/M_abs  # 14/3 ≈ 4.6667

M_plot = 1.0
l_plot = l_abs * scale_factor  # 1.5 * (14/3) = 7.0
print(f"\nEn unidades de la figura (M=1 en ejes):")
print(f"l_plot = {l_plot:.4f}")
print(f"Esto es l_plot/M_plot = {l_plot:.4f}")

# Parámetro μ para la figura
mu_plot = M_plot / l_plot  # 1/7 ≈ 0.142857
print(f"µ_plot = {mu_plot:.6f}")

# ============================================================================
# 1. ÓRBITA CIRCULAR (en unidades de figura)
# ============================================================================
r_circ_plot = l_plot  # 7.0
phi_circ = np.linspace(0, 2*np.pi, 300)
x_circ = r_circ_plot * np.cos(phi_circ)
y_circ = r_circ_plot * np.sin(phi_circ)

# ============================================================================
# 2. ÓRBITA DEL SEGUNDO TIPO (en unidades de figura)
# ============================================================================
# u3 en unidades de figura
u3_plot = 1/(2*M_plot) - 2/l_plot  # 0.5 - 2/7 = 0.5 - 0.2857 = 0.2143
r_ap_plot = 1/u3_plot  # ≈ 4.6667 (que es 1/M_abs, coherente)

print(f"\nEn unidades de figura:")
print(f"u3_plot = {u3_plot:.4f}")
print(f"r_ap_plot = {r_ap_plot:.4f}")

# Ángulo hasta la singularidad
phi_sing_plot = np.pi / np.sqrt(1 - 6*mu_plot)  # π/√(1-6/7) = π/√(1/7) = π√7 ≈ 8.31 rad
print(f"Ángulo afelio→singularidad: {phi_sing_plot:.4f} rad")

# Generar la órbita
phi_max_plot = phi_sing_plot * 0.99
phi_vals = np.linspace(0, phi_max_plot, 500)
theta = 0.5 * np.sqrt(1 - 6*mu_plot) * phi_vals
u_vals = 1/l_plot + (1/(2*M_plot) - 3/l_plot) * (1/np.cos(theta))**2
r_vals = 1/u_vals

# Convertir a cartesianas
x_vals = r_vals * np.cos(phi_vals)
y_vals = r_vals * np.sin(phi_vals)

# ============================================================================
# 3. HORIZONTE (r=2M) en unidades de figura
# ============================================================================
r_horizon_plot = 2*M_plot  # = 2.0
phi_hor = np.linspace(0, 2*np.pi, 100)
x_hor = r_horizon_plot * np.cos(phi_hor)
y_hor = r_horizon_plot * np.sin(phi_hor)

# ============================================================================
# 4. GRÁFICO CORREGIDO
# ============================================================================
plt.figure(figsize=(9, 9))

# Horizonte
plt.plot(x_hor, y_hor, 'k--', linewidth=1.5, alpha=0.7, label=f'Horizonte (r=2M)')

# Órbita circular (EXTERIOR)
plt.plot(x_circ, y_circ, 'b-', linewidth=2.5, label=f'Órbita circular (r={r_circ_plot:.1f}M)')

# Órbita del segundo tipo (INTERIOR, empieza en r_ap ≈ 4.67M)
plt.plot(x_vals, y_vals, 'r-', linewidth=2.5, label='Órbita del 2º tipo (e=0)')

# Puntos importantes
plt.plot(x_vals[0], y_vals[0], 'ro', markersize=10, label='Afelio (inicio)')
plt.plot(0, 0, 'k*', markersize=15, label='Singularidad (r=0)')

# Configuración
plt.axhline(0, color='gray', linestyle=':', linewidth=0.7, alpha=0.5)
plt.axvline(0, color='gray', linestyle=':', linewidth=0.7, alpha=0.5)
plt.gca().set_aspect('equal', adjustable='box')
plt.xlabel('x (M=3/14 en unidades físicas)', fontsize=12)
plt.ylabel('y (M=3/14 en unidades físicas)', fontsize=12)
plt.title('Figura 7a(e): Órbita circular y órbita del segundo tipo (e=0)\n' +
          f'Parámetros: l={l_abs}, M={M_abs:.4f} (absolutos)', fontsize=14)
plt.legend(loc='upper right', fontsize=10)
plt.grid(True, alpha=0.3)

# Límites: el círculo tiene radio 7, así que necesitamos eje hasta ~8
plt.xlim(-8, 8)
plt.ylim(-8, 8)

# Añadir anotaciones
circle_radius_physical = r_circ_plot * M_abs  # 7 * (3/14) = 1.5
plt.text(0.5, 7.5, f'Radio circular físico: {circle_radius_physical:.2f}', 
         fontsize=10, color='blue', bbox=dict(boxstyle="round,pad=0.3", facecolor="white", alpha=0.8))
plt.text(0.5, 6.8, f'Radio afelio físico: {r_ap_plot*M_abs:.2f}', 
         fontsize=10, color='red', bbox=dict(boxstyle="round,pad=0.3", facecolor="white", alpha=0.8))

plt.tight_layout()
plt.show()

print("\n" + "="*50)
print("RESUMEN:")
print("="*50)
print(f"1. Órbita circular: radio físico = {circle_radius_physical:.2f}")
print(f"2. Órbita del segundo tipo:")
print(f"   - Comienza en afelio: r = {r_ap_plot*M_abs:.2f} (físico)")
print(f"   - Da {phi_sing_plot/(2*np.pi):.2f} vueltas antes de singularidad")
print(f"3. Ambas órbitas están FUERA del horizonte (r=2M={2*M_abs:.2f})")
```
