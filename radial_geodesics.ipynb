## Las geodésicas en el espacio-tiempo de Schwarzschild; Las geodésicas tipo tiempo

$$ds=\sqrt{g_{ij}\frac{dx^{i}}{d\tau}\frac{dx^{j}}{d\tau} }d\tau, \ \quad \ s=-m\int \sqrt{g_{ij}\frac{dx^{i}}{d\tau}\frac{dx^{j}}{d\tau}  }d\tau , \ \quad \therefore \ \mathcal{L}=-m\sqrt{g_{ij}\frac{dx^{i}}{d\tau}\frac{dx^{j}}{d\tau}  }$$


$$
ds^2 = g_{ij}dx^{i} dy^{j} 
\rightarrow \ \ 2\mathcal{L}=g_{ij}\frac{dx^{i}}{d\tau}\frac{dx^{j}}{d\tau}
$$    

$$\mathcal{L}=\frac{1}{2}\left \{ \left(1 - \frac{2M}{r} \right)\dot{t}^{2} - \frac{\dot{r}^{2}}{1-\frac{2M}{r}} - r^2 \dot{\theta}^{2} -(r^2 \sin^{2} \theta)\dot{\phi}^2 \right \}  \left\{\begin{matrix}
 \dot{t}=\frac{dt}{d\tau}\\
 \dot{r}=\frac{dr}{d\tau}\\
 \dot{\theta}=\frac{d\theta}{d\tau} \\
 \dot{\phi}=\frac{d\phi}{d\tau}
\end{matrix}\right.$$

Momentos Canónicos:

$$p_t=\partial_{\dot{t}}\mathcal{L}=\left(1- \frac{2M}{r} \right)\dot{t}, \ \quad \ p_r=-\partial_{\dot{r}}\mathcal{L}=\left(1- \frac{2M}{r} \right)^{-1}\dot{r}, \ \quad \ p_\phi=-\partial_{\dot{\phi}}\mathcal{L}=\left(r^2 \sin^2 \theta \right)\dot{\phi}, \ \quad \ p_{\theta}=-\partial_{\theta}\mathcal{L} = r^2 \dot{\theta} $$


Por lo que el Hamiltoniano,


$$H=\sum_i p_i \dot{q}_i - \mathcal{L} = p_{t}\dot{t}-(p_r \dot{r} + p_{\theta} \dot{\theta} + p_{\phi} \dot{\phi}) - \mathcal{L}$$


$$\dot{t}=\frac{p_t}{\left(1-\frac{2M}{r} \right)}, \ \quad \ \dot{r}= \frac{p_r}{\left(1 - \frac{2M}{r} \right)^{-1}}, \ \quad \ \dot{\phi}= \frac{p_\phi}{r^2 \sin^2 \theta} , \ \quad \ \dot{\theta}= \frac{p_\theta}{r^2} $$

$$p_t \dot{t} = (1-\frac{2M}{r})\dot{t}^2 , \ \quad \ p_r\dot{r}= \left(1- \frac{2M}{r} \right)^{-1} \dot{r}^2, \ \quad \ p_\theta \dot{\theta}= r^2 \dot{\theta}^2, \ \quad \ p_\phi \dot{\phi} = (r^2\sin^2 \theta)\dot{\phi}^2 $$

$$ \therefore \ H = \mathcal{L}$$

$$
\mathrm{d} s^2=g_{i j} \mathrm{~d} x^i \mathrm{~d} x^j,
$$

can be derived from the Lagrangian

$$
2 \mathscr{L}=g_{i j} \frac{\mathrm{~d} x^i}{\mathrm{~d} \tau} \frac{\mathrm{~d} x^j}{\mathrm{~d} \tau},
$$

where $\tau$ is some affine parameter along the geodesic. For time-like geodesics, $\tau$ may be identified with the proper time, $s$, of the particle describing the geodesic.

For the Schwarzschild space-time, the Lagrangian is

$$
\mathscr{L}=\frac{1}{2}\left[\left(1-\frac{2 M}{r}\right) \dot{t}^2-\frac{\dot{r}^2}{1-2 M / r}-r^2 \dot{\theta}^2-\left(r^2 \sin ^2 \theta\right) \dot{\varphi}^2\right],
$$

where the dot denotes differentiation with respect to $\tau$. The corresponding canonical momenta are

$$
\left.\begin{array}{l}
p_t=\frac{\partial \mathscr{L}}{\partial \dot{t}}=\left(1-\frac{2 M}{r}\right) \dot{t}, \quad p_{\varphi}=-\frac{\partial \mathscr{L}}{\partial \dot{\varphi}}=\left(r^2 \sin ^2 \theta\right) \dot{\varphi}, \\
p_r=-\frac{\partial \mathscr{L}}{\partial \dot{r}}=\left(1-\frac{2 M}{r}\right)^{-1} \dot{r}, \quad \text { and } \quad p_\theta=-\frac{\partial \mathscr{L}}{\partial \theta}=r^2 \theta .
\end{array}\right\}
$$


The resulting Hamiltonian is

$$
\mathscr{H}=p_t \dot{t}-\left(p_r \dot{r}+p_\theta \dot{\theta}+p_{\varphi} \dot{\varphi}\right)-\mathscr{L}=\mathscr{L} .
$$




```python
import numpy as np
import matplotlib.pyplot as plt

# Parámetro del agujero negro
M = 1.0

# Valores de E entre 0 y 1 con paso 0.1
E_values = np.arange(0.0, 1.0, 0.1)

# Rango del parámetro eta
eta = np.linspace(0, np.pi, 600)

plt.figure(figsize=(10, 6))

for E in E_values:
    # r(eta) según la ecuación paramétrica
    r = (M / (1 - E**2)) * (1 + np.cos(eta))
    
    # Dibujar curva r(eta)
    plt.plot(eta, r, label=f"E={E:.1f}")
    
    # Valor eta_H del cruce del horizonte
    eta_H = 2 * np.arcsin(E)
    
    # Dibujar línea vertical indicando eta_H
    plt.axvline(eta_H, linestyle="--", alpha=0.3)

plt.xlabel("η")
plt.ylabel("r(η)")
plt.title("Curvas r(η) para E = 0.0, 0.1, ..., 0.9  (M=1)")
plt.grid(True)
plt.legend()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Constantes
M = 1.0
r0 = 6.0 * M  # posición inicial
eta0 = 0.0    # parámetro eta inicial (en reposo, dr/dtau=0)

# Para una caída radial desde el reposo en r0, E^2 = 1 - 2M/r0
E = np.sqrt(1 - 2*M/r0)

# Ecuación de la órbita radial en coordenadas de Schwarzschild
# r = (r0/2) * (1 + cos(eta))
# t en función de eta (para eta desde 0 hasta pi)
# tau en función de eta

# Generar valores de eta desde 0 hasta pi
eta = np.linspace(0, np.pi, 500)

# Radio en función de eta
r = 0.5 * r0 * (1 + np.cos(eta))

# Tiempo propio tau (fórmula conocida para caída radial)
# dtau/deta = sqrt(r0^3/(8M)) * (1 + cos(eta))
tau = np.sqrt(r0**3 / (8*M)) * (eta + np.sin(eta))
# Ajustamos para que empiece en 0
tau = tau - tau[0]

# Tiempo de Schwarzschild t (fórmula más compleja, pero simplificamos)
# Para caída desde reposo, E^2 = 1 - 2M/r0
# La fórmula completa (Chandrasekhar) da:
# t = sqrt(r0^3/(2M)) * [1/2*(eta + sin(eta)) + (1-E^2)*eta] 
#     + 2M * log((tan(eta_H/2) + tan(eta/2))/(tan(eta_H/2) - tan(eta/2)))
# donde cos^2(eta_H/2) = 2M/r0

eta_H = 2 * np.arccos(np.sqrt(2*M/r0))
t_prefactor = np.sqrt(r0**3 / (2*M))

# Calculamos t
t = t_prefactor * (0.5*(eta + np.sin(eta)) + (1 - E**2)*eta)
t_log_term = 2*M * np.log(np.abs((np.tan(eta_H/2) + np.tan(eta/2)) / 
                                 (np.tan(eta_H/2) - np.tan(eta/2))))
t = t + t_log_term
# Ajustamos para que empiece en 0
t = t - t[0]

# Crear la figura
plt.figure(figsize=(10, 6))

# Gráfico de r/M vs t/M y τ/M
plt.plot(t/M, r/M, 'b-', linewidth=2, label='t, Schwarzschild time')
plt.plot(tau/M, r/M, 'r-', linewidth=2, label='τ, proper time')

# Configuración del gráfico
plt.xlabel('time/M', fontsize=12)
plt.ylabel('r/M', fontsize=12)
plt.title('Radial geodesic: particle falling from rest at r = 6M', fontsize=14)
plt.grid(True, alpha=0.3)
plt.legend(fontsize=12)
plt.axhline(y=2, color='k', linestyle='--', alpha=0.5, label='Horizon (r=2M)')
plt.axhline(y=0, color='k', linestyle='-', alpha=0.5, label='Singularity (r=0)')
plt.xlim(0, 30)  # Como en la figura original
plt.ylim(0, 6.5)  # Desde singularidad hasta un poco más de 6M

# Añadir texto descriptivo
plt.text(15, 5, 'Particle starts at rest at r=6M', fontsize=10)
plt.text(15, 4.5, 'and falls toward singularity', fontsize=10)

plt.legend(loc='upper right')
plt.tight_layout()
plt.show()

# También podemos graficar por separado para comparar con la figura
fig, ax = plt.subplots(figsize=(10, 6))

# Líneas más gruesas como en la figura original
ax.plot(t/M, r/M, 'b-', linewidth=3, label='t, Schwarzschild time')
ax.plot(tau/M, r/M, 'r-', linewidth=3, label='τ, proper time')

# Configuración estética similar a la figura
ax.set_xlabel('time/M', fontsize=14)
ax.set_ylabel('r/M', fontsize=14)
ax.set_title('Radial Timelike Geodesic in Schwarzschild Spacetime', fontsize=16)
ax.grid(True, alpha=0.2)
ax.set_xlim(0, 30)
ax.set_ylim(0, 6.5)

# Líneas horizontales para horizonte y singularidad
ax.axhline(y=2, color='black', linestyle=':', linewidth=1, alpha=0.7)
ax.text(1, 2.1, 'Horizon (r=2M)', fontsize=10, alpha=0.7)
ax.axhline(y=0, color='black', linestyle='-', linewidth=1, alpha=0.5)
ax.text(1, 0.2, 'Singularity (r=0)', fontsize=10, alpha=0.7)

ax.legend(fontsize=12, loc='upper right')
plt.tight_layout()
plt.show()
```

## El potencial efectivo apropiado para trayectorias tipo-tiempo 

$$ V_{eff}=\left(1-\frac{2M}{r} \right)\left(1+\frac{L^2}{r^2}\right ) $$

Los mínimos en el potencial corresponden a las órbitas circulares estables mientras que los máximos corresponden a órbitas circulares inestables. Los puntos de inflexión es donde la orbita circular estable ocurre.

```python
import numpy as np
import matplotlib.pyplot as plt

# Constants
M = 1  # Mass of the central object, can be scaled as 1 for simplicity

# Define the effective potential function
def V_eff(r, L):
    return (1 - 2*M/r) * (1 + L**2 / r**2)

# Define the range for r (radial distance)
r_vals = np.linspace(2.5,25, 500)  # Full range of r

# Different values of L (angular momentum) to plot
L_vals = [4.33, 4, 3.75, 3.464]


# Each tuple contains (V_constant, r_start, r_end)
horizontal_lines = [
    (0.942, 12, 20),  # Line at V_eff = 0.94, for r from 10 to 20
    (0.93, 10, 15),   # Line at V_eff = 1.0, for r from 12 to 18
    (0.918, 7.2, 13),  # Line at V_eff = 0.98, for r from 15 to 22
]

# Plot the effective potential for each L value
plt.figure(figsize=(8, 6))
for L in L_vals:
    V = V_eff(r_vals, L)
    plt.plot(r_vals, V, label=f"L/M = {L:.2f}")

    # Find the minimum of the effective potential within the range 5 <= r <= 10
    # Restricting r_vals to the range 5 to 10 for the minimum calculation
    r_range = r_vals[(r_vals >= 5) & (r_vals <= 20)]
    V_range = V[(r_vals >= 5) & (r_vals <= 20)]
    
    # Finding the minimum value of V within the restricted range
    r_min_idx = np.argmin(V_range)
    r_min = r_range[r_min_idx]
    V_min = V_range[r_min_idx]
    
    # Print the minimum values
    print(f"Minimum for L/M = {L:.2f}: r = {r_min:.4f}, V_eff = {V_min:.4f}")
    
    # Mark the minimum with a red star
    plt.scatter(r_min, V_min, color='red', zorder=5, marker='*')  # Red star for the minimum

for V_const, r_start, r_end in horizontal_lines:
    plt.plot([r_start, r_end], [V_const, V_const], color='black', linestyle='dashed', label=f"V_eff = {V_const} (r: {r_start}-{r_end})")



# Adding labels and a legend
plt.axhline(0, color='black', linewidth=0.5)
plt.axvline(0, color='black', linewidth=0.5)
plt.title('Effective Potential for Time-like Trajectories')
plt.ylim(0.8, 1.1)
plt.xlabel('r/M')
plt.ylabel('V_eff')
plt.legend()
plt.grid(True)

# Show the plot
plt.show()
```
