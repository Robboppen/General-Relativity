## Las geodésicas en el espacio-tiempo de Schwarzschild; Las geodésicas tipo tiempo

$$ds=\sqrt{g_{ij}\frac{dx^{i}}{d\tau}\frac{dx^{j}}{d\tau} }d\tau, \ \quad \ s=-m\int \sqrt{g_{ij}\frac{dx^{i}}{d\tau}\frac{dx^{j}}{d\tau}  }d\tau , \ \quad \therefore \ \mathcal{L}=-m\sqrt{g_{ij}\frac{dx^{i}}{d\tau}\frac{dx^{j}}{d\tau}  }$$


$$
ds^2 = g_{ij}dx^{i} dy^{j} 
\rightarrow \ \ 2\mathcal{L}=g_{ij}\frac{dx^{i}}{d\tau}\frac{dx^{j}}{d\tau}
$$    

$$\mathcal{L}=\frac{1}{2}\left \{ \left(1 - \frac{2M}{r} \right)\dot{t}^{2} - \frac{\dot{r}^{2}}{1-\frac{2M}{r}} - r^2 \dot{\theta}^{2} -(r^2 \sin^{2} \theta)\dot{\phi}^2 \right \}  \left\{\begin{matrix}
 \dot{t}=\frac{dt}{d\tau}\\
 \dot{r}=\frac{dr}{d\tau}\\
 \dot{\theta}=\frac{d\theta}{d\tau} \\
 \dot{\phi}=\frac{d\phi}{d\tau}
\end{matrix}\right.$$

Momentos Canónicos:

$$p_t=\partial_{\dot{t}}\mathcal{L}=\left(1- \frac{2M}{r} \right)\dot{t}, \ \quad \ p_r=-\partial_{\dot{r}}\mathcal{L}=\left(1- \frac{2M}{r} \right)^{-1}\dot{r}, \ \quad \ p_\phi=-\partial_{\dot{\phi}}\mathcal{L}=\left(r^2 \sin^2 \theta \right)\dot{\phi}, \ \quad \ p_{\theta}=-\partial_{\theta}\mathcal{L} = r^2 \dot{\theta} $$


Por lo que el Hamiltoniano,


$$H=\sum_i p_i \dot{q}_i - \mathcal{L} = p_{t}\dot{t}-(p_r \dot{r} + p_{\theta} \dot{\theta} + p_{\phi} \dot{\phi}) - \mathcal{L}$$


$$\dot{t}=\frac{p_t}{\left(1-\frac{2M}{r} \right)}, \ \quad \ \dot{r}= \frac{p_r}{\left(1 - \frac{2M}{r} \right)^{-1}}, \ \quad \ \dot{\phi}= \frac{p_\phi}{r^2 \sin^2 \theta} , \ \quad \ \dot{\theta}= \frac{p_\theta}{r^2} $$

$$p_t \dot{t} = (1-\frac{2M}{r})\dot{t}^2 , \ \quad \ p_r\dot{r}= \left(1- \frac{2M}{r} \right)^{-1} \dot{r}^2, \ \quad \ p_\theta \dot{\theta}= r^2 \dot{\theta}^2, \ \quad \ p_\phi \dot{\phi} = (r^2\sin^2 \theta)\dot{\phi}^2 $$

$$ \therefore \ H = \mathcal{L}$$

$$
\mathrm{d} s^2=g_{i j} \mathrm{~d} x^i \mathrm{~d} x^j,
$$

can be derived from the Lagrangian

$$
2 \mathscr{L}=g_{i j} \frac{\mathrm{~d} x^i}{\mathrm{~d} \tau} \frac{\mathrm{~d} x^j}{\mathrm{~d} \tau},
$$

where $\tau$ is some affine parameter along the geodesic. For time-like geodesics, $\tau$ may be identified with the proper time, $s$, of the particle describing the geodesic.

For the Schwarzschild space-time, the Lagrangian is

$$
\mathscr{L}=\frac{1}{2}\left[\left(1-\frac{2 M}{r}\right) \dot{t}^2-\frac{\dot{r}^2}{1-2 M / r}-r^2 \dot{\theta}^2-\left(r^2 \sin ^2 \theta\right) \dot{\varphi}^2\right],
$$

where the dot denotes differentiation with respect to $\tau$. The corresponding canonical momenta are

$$
\left.\begin{array}{l}
p_t=\frac{\partial \mathscr{L}}{\partial \dot{t}}=\left(1-\frac{2 M}{r}\right) \dot{t}, \quad p_{\varphi}=-\frac{\partial \mathscr{L}}{\partial \dot{\varphi}}=\left(r^2 \sin ^2 \theta\right) \dot{\varphi}, \\
p_r=-\frac{\partial \mathscr{L}}{\partial \dot{r}}=\left(1-\frac{2 M}{r}\right)^{-1} \dot{r}, \quad \text { and } \quad p_\theta=-\frac{\partial \mathscr{L}}{\partial \theta}=r^2 \theta .
\end{array}\right\}
$$


The resulting Hamiltonian is

$$
\mathscr{H}=p_t \dot{t}-\left(p_r \dot{r}+p_\theta \dot{\theta}+p_{\varphi} \dot{\varphi}\right)-\mathscr{L}=\mathscr{L} .
$$




```python
import numpy as np
import matplotlib.pyplot as plt

# Parámetro del agujero negro
M = 1.0

# Valores de E entre 0 y 1 con paso 0.1
E_values = np.arange(0.0, 1.0, 0.1)

# Rango del parámetro eta
eta = np.linspace(0, np.pi, 600)

plt.figure(figsize=(10, 6))

for E in E_values:
    # r(eta) según la ecuación paramétrica
    r = (M / (1 - E**2)) * (1 + np.cos(eta))
    
    # Dibujar curva r(eta)
    plt.plot(eta, r, label=f"E={E:.1f}")
    
    # Valor eta_H del cruce del horizonte
    eta_H = 2 * np.arcsin(E)
    
    # Dibujar línea vertical indicando eta_H
    plt.axvline(eta_H, linestyle="--", alpha=0.3)

plt.xlabel("η")
plt.ylabel("r(η)")
plt.title("Curvas r(η) para E = 0.0, 0.1, ..., 0.9  (M=1)")
plt.grid(True)
plt.legend()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Constantes
M = 1.0
r0 = 6.0 * M  # posición inicial
eta0 = 0.0    # parámetro eta inicial (en reposo, dr/dtau=0)

# Para una caída radial desde el reposo en r0, E^2 = 1 - 2M/r0
E = np.sqrt(1 - 2*M/r0)

# Ecuación de la órbita radial en coordenadas de Schwarzschild
# r = (r0/2) * (1 + cos(eta))
# t en función de eta (para eta desde 0 hasta pi)
# tau en función de eta

# Generar valores de eta desde 0 hasta pi
eta = np.linspace(0, np.pi, 500)

# Radio en función de eta
r = 0.5 * r0 * (1 + np.cos(eta))

# Tiempo propio tau (fórmula conocida para caída radial)
# dtau/deta = sqrt(r0^3/(8M)) * (1 + cos(eta))
tau = np.sqrt(r0**3 / (8*M)) * (eta + np.sin(eta))
# Ajustamos para que empiece en 0
tau = tau - tau[0]

# Tiempo de Schwarzschild t (fórmula más compleja, pero simplificamos)
# Para caída desde reposo, E^2 = 1 - 2M/r0
# La fórmula completa (Chandrasekhar) da:
# t = sqrt(r0^3/(2M)) * [1/2*(eta + sin(eta)) + (1-E^2)*eta] 
#     + 2M * log((tan(eta_H/2) + tan(eta/2))/(tan(eta_H/2) - tan(eta/2)))
# donde cos^2(eta_H/2) = 2M/r0

eta_H = 2 * np.arccos(np.sqrt(2*M/r0))
t_prefactor = np.sqrt(r0**3 / (2*M))

# Calculamos t
t = t_prefactor * (0.5*(eta + np.sin(eta)) + (1 - E**2)*eta)
t_log_term = 2*M * np.log(np.abs((np.tan(eta_H/2) + np.tan(eta/2)) / 
                                 (np.tan(eta_H/2) - np.tan(eta/2))))
t = t + t_log_term
# Ajustamos para que empiece en 0
t = t - t[0]

# Crear la figura
plt.figure(figsize=(10, 6))

# Gráfico de r/M vs t/M y τ/M
plt.plot(t/M, r/M, 'b-', linewidth=2, label='t, Schwarzschild time')
plt.plot(tau/M, r/M, 'r-', linewidth=2, label='τ, proper time')

# Configuración del gráfico
plt.xlabel('time/M', fontsize=12)
plt.ylabel('r/M', fontsize=12)
plt.title('Radial geodesic: particle falling from rest at r = 6M', fontsize=14)
plt.grid(True, alpha=0.3)
plt.legend(fontsize=12)
plt.axhline(y=2, color='k', linestyle='--', alpha=0.5, label='Horizon (r=2M)')
plt.axhline(y=0, color='k', linestyle='-', alpha=0.5, label='Singularity (r=0)')
plt.xlim(0, 30)  # Como en la figura original
plt.ylim(0, 6.5)  # Desde singularidad hasta un poco más de 6M

# Añadir texto descriptivo
plt.text(15, 5, 'Particle starts at rest at r=6M', fontsize=10)
plt.text(15, 4.5, 'and falls toward singularity', fontsize=10)

plt.legend(loc='upper right')
plt.tight_layout()
plt.show()

# También podemos graficar por separado para comparar con la figura
fig, ax = plt.subplots(figsize=(10, 6))

# Líneas más gruesas como en la figura original
ax.plot(t/M, r/M, 'b-', linewidth=3, label='t, Schwarzschild time')
ax.plot(tau/M, r/M, 'r-', linewidth=3, label='τ, proper time')

# Configuración estética similar a la figura
ax.set_xlabel('time/M', fontsize=14)
ax.set_ylabel('r/M', fontsize=14)
ax.set_title('Radial Timelike Geodesic in Schwarzschild Spacetime', fontsize=16)
ax.grid(True, alpha=0.2)
ax.set_xlim(0, 30)
ax.set_ylim(0, 6.5)

# Líneas horizontales para horizonte y singularidad
ax.axhline(y=2, color='black', linestyle=':', linewidth=1, alpha=0.7)
ax.text(1, 2.1, 'Horizon (r=2M)', fontsize=10, alpha=0.7)
ax.axhline(y=0, color='black', linestyle='-', linewidth=1, alpha=0.5)
ax.text(1, 0.2, 'Singularity (r=0)', fontsize=10, alpha=0.7)

ax.legend(fontsize=12, loc='upper right')
plt.tight_layout()
plt.show()
```

## El potencial efectivo apropiado para trayectorias tipo-tiempo 

$$ V_{eff}=\left(1-\frac{2M}{r} \right)\left(1+\frac{L^2}{r^2}\right ) $$

Los mínimos en el potencial corresponden a las órbitas circulares estables mientras que los máximos corresponden a órbitas circulares inestables. Los puntos de inflexión es donde la orbita circular estable ocurre.

```python
import numpy as np
import matplotlib.pyplot as plt

# Constants
M = 1  # Mass of the central object, can be scaled as 1 for simplicity

# Define the effective potential function
def V_eff(r, L):
    return (1 - 2*M/r) * (1 + L**2 / r**2)

# Define the range for r (radial distance)
r_vals = np.linspace(2.5,25, 500)  # Full range of r

# Different values of L (angular momentum) to plot
L_vals = [4.33, 4, 3.75, 3.464]


# Each tuple contains (V_constant, r_start, r_end)
horizontal_lines = [
    (0.942, 12, 20),  # Line at V_eff = 0.94, for r from 10 to 20
    (0.93, 10, 15),   # Line at V_eff = 1.0, for r from 12 to 18
    (0.918, 7.2, 13),  # Line at V_eff = 0.98, for r from 15 to 22
]

# Plot the effective potential for each L value
plt.figure(figsize=(8, 6))
for L in L_vals:
    V = V_eff(r_vals, L)
    plt.plot(r_vals, V)

    # Find the minimum of the effective potential within the range 5 <= r <= 10
    # Restricting r_vals to the range 5 to 10 for the minimum calculation
    r_range = r_vals[(r_vals >= 5) & (r_vals <= 20)]
    V_range = V[(r_vals >= 5) & (r_vals <= 20)]
    
    # Finding the minimum value of V within the restricted range
    r_min_idx = np.argmin(V_range)
    r_min = r_range[r_min_idx]
    V_min = V_range[r_min_idx]
    
    # Print the minimum values
    print(f"Minimum for L/M = {L:.2f}: r = {r_min:.4f}, V_eff = {V_min:.4f}")
    
    # Mark the minimum with a red star
    plt.scatter(r_min, V_min, color='red', zorder=5, marker='*')  # Red star for the minimum

for V_const, r_start, r_end in horizontal_lines:
    plt.plot([r_start, r_end], [V_const, V_const], color='black', linestyle='dashed')



# Adding labels and a legend
plt.axhline(0, color='black', linewidth=0.5)
plt.axvline(0, color='black', linewidth=0.5)
plt.title('Potencial Efectivo para órbitas tipo-tiempo.')
plt.ylim(0.8, 1.1)
plt.xlabel('r/M')
plt.ylabel('V_eff')
plt.legend()
plt.grid(True)

# Show the plot
plt.show()
```

$$\left(\frac{d \chi }{d \phi} \right)^2 = 1-2\mu(3+e \cos \chi) = (1 - 6\mu +2e\mu ) - 4\mu e \cos^2 \frac{1}{2}\chi$$

$$\pm \frac{d \chi}{d \phi} = (1-6\mu + 2\mu e)^{1/2}(1-k^2 \cos^2 \frac{1}{2}\chi)^{1/2}, \quad \ k^2 = \frac{4 \mu e}{1- 6\mu + 2\mu e} $$


$$d\phi = \pm \frac{d \chi}{\sqrt{A}\sqrt{1-k^2\sin^2 \frac{\chi}{2}}}, \quad \gamma=\frac{1}{2}\chi $$

Afelio $\chi = \pi \rightarrow \gamma = \frac{\pi}{2} \rightarrow \psi = 0$ 

$$F(\psi, k)=\int^\psi_0 \frac{d\gamma}{\sqrt{1-k^2\sin^2 \gamma}}, \quad \ \psi = \frac{1}{2}(\pi - \chi), \quad \psi= \frac{2}{(1-6\mu + 2e\mu)^{1/2}} F(\frac{1}{2}\pi - \frac{1}{2}\chi, k )$$

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.special import ellipk, ellipj, ellipkinc
from scipy.integrate import quad
from scipy.interpolate import interp1d

# Configuración de matplotlib para mejor calidad
plt.rcParams.update({
    'font.size': 12,
    'figure.figsize': (14, 10),
    'axes.grid': True,
    'grid.alpha': 0.3,
    'axes.axisbelow': True
})

# ============================================================================
# PARÁMETROS (valores del texto, figura 7a)
# ============================================================================
# Valores para las tres subfiguras (a), (b), (c)
param_sets = [
    {'e': 0.5, 'l': 11.0, 'M': 3/14, 'label': '(a) e=0.5, l=11, M=3/14', 'revoluciones': 3},
    {'e': 0.5, 'l': 7.5, 'M': 3/14, 'label': '(b) e=0.5, l=7.5, M=3/14', 'revoluciones': 3},
    {'e': 0.5, 'l': 3.0, 'M': 3/14, 'label': '(c) e=0.5, l=3, M=3/14', 'revoluciones': 3}
]

# ============================================================================
# FUNCIONES AUXILIARES
# ============================================================================
def calcular_mu(M, l):
    """Calcula μ = M/l"""
    return M / l

def calcular_A(mu, e):
    """Calcula A = 1 - 6μ + 2μe"""
    return 1 - 6*mu + 2*mu*e

def calcular_k2(mu, e, A):
    """Calcula k² = 4μe / A"""
    return 4*mu*e / A if A != 0 else 1.0

def phi_from_chi(chi, A, k2):
    """
    Calcula φ(χ) usando la ecuación (132):
    φ = 2/√A * F(π/2 - χ/2, k)
    donde F es la integral elíptica incompleta de primera especie
    """
    if A <= 0:
        return np.nan
    
    sqrt_A = np.sqrt(A)
    k = np.sqrt(k2)
    
    # ψ = π/2 - χ/2
    psi = np.pi/2 - chi/2
    
    # Usamos scipy.special.ellipkinc para la integral elíptica incompleta
    # ellipkinc(u, m) donde m = k²
    F_value = ellipkinc(psi, k2)
    
    return 2 * F_value / sqrt_A

def chi_from_phi_total(phi_total, A, k2, chi_range=(0, 2*np.pi), num_points=2000):
    """
    Encuentra χ para un φ dado, considerando múltiples revoluciones
    """
    chi_vals = np.linspace(chi_range[0], chi_range[1], num_points)
    phi_vals = phi_from_chi(chi_vals, A, k2)
    
    # Desenrollar φ para que sea monótono creciente
    phi_unwrapped = np.unwrap(phi_vals)
    
    # Interpolar
    f_interp = interp1d(phi_unwrapped, chi_vals, kind='cubic', 
                       fill_value='extrapolate', bounds_error=False)
    
    return f_interp(phi_total)

def orbit_r_phi_multiple(params, num_revolutions=3, points_per_revolution=500):
    """
    Calcula la órbita r(φ) para múltiples revoluciones
    """
    e = params['e']
    l = params['l']
    M = params['M']
    
    # Calcular parámetros intermedios
    mu = calcular_mu(M, l)
    A = calcular_A(mu, e)
    k2 = calcular_k2(mu, e, A)
    
    if A <= 0 or k2 < 0:
        print(f"Parámetros inválidos: A={A:.4f}, k²={k2:.4f}")
        return np.array([]), np.array([])
    
    # Calcular φ total para una revolución (de afelio a afelio)
    # χ de π (afelio) a -π (siguiente afelio)
    chi_start = np.pi
    chi_end = -np.pi
    
    chi_vals_rev = np.linspace(chi_start, chi_end, points_per_revolution)
    phi_vals_rev = phi_from_chi(chi_vals_rev, A, k2)
    
    # φ total para una revolución
    phi_per_revolution = phi_vals_rev[-1] - phi_vals_rev[0]
    
    print(f"  Ángulo por revolución: Δφ = {phi_per_revolution:.4f} rad")
    
    # Generar múltiples revoluciones
    all_chi = []
    all_phi = []
    
    for rev in range(num_revolutions):
        # Para cada revolución, desplazamos χ
        chi_offset = rev * 2 * np.pi
        chi_vals_adj = chi_vals_rev - chi_offset
        
        phi_offset = rev * phi_per_revolution
        phi_vals_adj = phi_vals_rev - phi_vals_rev[0] + phi_offset
        
        all_chi.extend(chi_vals_adj)
        all_phi.extend(phi_vals_adj)
    
    all_chi = np.array(all_chi)
    all_phi = np.array(all_phi)
    
    # Calcular r = 1/u, con u = (1 + e cosχ)/l
    r_vals = l / (1 + e * np.cos(all_chi))
    
    # Ajustar φ para empezar en 0
    phi_min = all_phi[0]
    all_phi_adj = all_phi - phi_min
    
    return r_vals, all_phi_adj, phi_per_revolution

# ============================================================================
# CÁLCULO Y GRÁFICA
# ============================================================================
def plot_orbits_multiple(param_sets):
    """
    Grafica múltiples órbitas con varias revoluciones
    """
    fig, axes = plt.subplots(1, 3, figsize=(18, 6))
    
    for idx, (params, ax) in enumerate(zip(param_sets, axes)):
        e = params['e']
        l = params['l']
        M = params['M']
        label = params['label']
        num_revolutions = params.get('revoluciones', 3)
        
        print(f"\nCalculando órbita {label}:")
        print(f"  e={e}, l={l}, M={M:.4f}, revoluciones={num_revolutions}")
        
        # Calcular órbita con múltiples revoluciones
        r_vals, phi_vals, phi_per_rev = orbit_r_phi_multiple(
            params, num_revolutions=num_revolutions, points_per_revolution=500
        )
        
        if len(r_vals) == 0:
            print(f"  ¡Advertencia: parámetros inválidos para órbita!")
            continue
        
        # Calcular parámetros para mostrar
        mu = calcular_mu(M, l)
        A = calcular_A(mu, e)
        k2 = calcular_k2(mu, e, A)
        
        print(f"  μ={mu:.4f}, A={A:.4f}, k²={k2:.4f}")
        print(f"  Ángulo por revolución: {phi_per_rev:.4f} rad")
        
        # Precesión por revolución
        precession_per_rev = phi_per_rev - 2*np.pi
        print(f"  Precesión por revolución: {precession_per_rev:.6f} rad = {precession_per_rev*180/np.pi:.4f}°")
        
        # Perihelio y afelio
        r_peri = l / (1 + e)
        r_ap = l / (1 - e) if e < 1 else np.inf
        
        # Radio del horizonte (r=2M) y órbita circular crítica (r=3M)
        r_horizon = 2*M
        r_critical = 3*M
        
        # Convertir a coordenadas cartesianas
        x_vals = r_vals * np.cos(phi_vals)
        y_vals = r_vals * np.sin(phi_vals)
        
       
        # Graficar segmentos con color progresivo
        for i in range(len(x_vals)-1):
            ax.plot(x_vals[i:i+2], y_vals[i:i+2],'b-', 
                   linewidth=1.5, alpha=0.8)
            ax.plot(0, 0, 'ko', markersize=7)
        
        
        # Configurar ejes
        ax.set_xlabel('x/M', fontsize=14)
        ax.set_ylabel('y/M', fontsize=14)
        ax.set_title(f'{label}\nμ={mu:.3f}, k²={k2:.3f}\n{num_revolutions} revoluciones', 
                    fontsize=14, pad=15)
        ax.set_aspect('equal', 'box')
        
        # Límites de gráfica
        max_r = max(r_vals) * 1.2
        min_r = min(r_vals) * 0.8
        lim = max(abs(max_r), abs(min_r))
        ax.set_xlim(-lim, lim)
        ax.set_ylim(-lim, lim)
        
        # Cuadrícula
        ax.grid(True, alpha=0.2, linestyle='--')
        
        # Leyenda
        ax.legend(loc='upper right', fontsize=8, framealpha=0.9)
    
    plt.suptitle('Órbitas de Tipo Tiempo en Schwarzschild - Múltiples Revoluciones\n', 
                 fontsize=16, y=1.02)
    plt.tight_layout()
    plt.show()
    
    return fig

# ============================================================================
# EJECUCIÓN PRINCIPAL
# ============================================================================
if __name__ == "__main__":
    print("=" * 70)
    print("Simulación de Órbitas Relativistas - Múltiples Revoluciones")
    print("Figura 7a (a,b,c) extendida con precesión del perihelio")
    print("=" * 70)
    
    # Graficar las tres órbitas con múltiples revoluciones
    fig = plot_orbits_multiple(param_sets)
    
    # Cálculo adicional: comparación con aproximación post-newtoniana
    print("\n" + "=" * 70)
    print("Comparación con aproximación post-newtoniana (Δφ = 6πM/l):")
    
    for params in param_sets:
        e = params['e']
        l = params['l']
        M = params['M']
        
        # Aproximación post-newtoniana
        delta_phi_approx = 6 * np.pi * M / l
        delta_phi_deg_approx = delta_phi_approx * 180 / np.pi
        
        print(f"  {params['label']}:")
        print(f"    Δφ_approx = {delta_phi_approx:.6f} rad = {delta_phi_deg_approx:.4f}° por revolución")
```

# El caso donde $2\mu (3+e)=1$

El perihelio $r_p$ y el afelio $r_{ap}$ estan dadas por:

$$r_p = \frac{l}{1+e}=2M\frac{3+e}{1+e}, \quad \ r_{ap}=2M \frac{3+e}{1-e}$$

El perihelio para estas orbitas estan restricted por el rango: $4M \leq r_p < 6M$

También para estas orbitas:

$$\frac{L^2}{M^2}= 4 \frac{(3+e)^2}{(3-e)(1+e)}, \quad \ 1-E^2 = \frac{1-e^2}{9-E^2} $$


El modulo $k$ de la integral elíptica se vuelve 1 por lo que tenemos estas ecuaciones


$$ \left(\frac{d \chi}{d \phi} \right)^2 = 4\mu e \sin^2 \frac{1}{2}\chi, \quad  \frac{d \chi }{d \phi} = -2(\mu e)^{1/2} \sin \frac{1}{2}\chi$$

Así,

$$\phi = - \frac{1}{\sqrt{\mu e}} lg \left(\tan \frac{1}{4} \chi \right)$$

```python
import numpy as np
import matplotlib.pyplot as plt

# ======================
# CONSTANTES Y PARÁMETROS
# ======================
M = 1.0
e = 0.5  # excentricidad para figura 7a(d)
mu = 1 / (2 * (3 + e))  # 2*mu*(3+e) = 1
l = M / mu

# Radio del círculo crítico
r_critico = l / (1 + e)  # Este es el círculo al que se aproximan las órbitas

print(f"Parámetros:")
print(f"M = {M}")
print(f"e = {e}")
print(f"μ = {mu:.4f}")
print(f"l = {l:.4f}")
print(f"Radio crítico r_c = l/(1+e) = {r_critico:.4f}")
print(f"Horizonte r = 2M = {2*M}")
print(f"Radio de última órbita estable en Schwarzschild = {6*M}")

# ======================
# FUNCIONES PARA LAS ECUACIONES
# ======================

def orbita_primer_tipo(chi_values):
    """
    Ecuación (150): φ = -1/sqrt(mu*e) * log(tan(χ/4))
    Para órbitas de primer tipo que se aproximan al círculo r_c desde fuera.
    χ va desde un valor inicial hasta 0 (cuando se acerca al círculo).
    """
    phi = -1 / np.sqrt(mu * e) * np.log(np.tan(chi_values / 4))
    
    # Radio para órbitas de primer tipo: ecuación (124) con χ como variable
    # En el caso especial 2μ(3+e)=1, hay que usar la relación apropiada
    # Según el texto: u = (1/l)(1 + e cos χ) para órbitas de primer tipo
    u = (1/l) * (1 + e * np.cos(chi_values))
    r = 1 / u
    
    return phi, r

def orbita_segundo_tipo(xi_values):
    """
    Ecuación (170): φ = -1/sqrt(mu*e) * log(tan(ξ/4))
    Para órbitas de segundo tipo que parten del círculo r_c hacia la singularidad.
    ξ va desde 0 (en el círculo) hasta π (en la singularidad).
    """
    phi = -1 / np.sqrt(mu * e) * np.log(np.tan(xi_values / 4))
    
    # Radio para órbitas de segundo tipo: ecuación (167)
    # u = (1/l)(1 + e + 2e tan²(ξ/2))
    u = (1/l) * (1 + e + 2 * e * np.tan(xi_values/2)**2)
    r = 1 / u
    
    return phi, r

# ======================
# GENERACIÓN DE DATOS
# ======================

# ----- Órbita de primer tipo (se aproxima al círculo desde fuera) -----
# χ comienza en un valor (lejos) y se acerca a 0 (círculo)
# Usamos valores de χ que van desde ~π/2 hasta cerca de 0
chi_min = 0.01  # cerca del círculo
chi_max = 1.5   # lejos del círculo (valor inicial)
chi_vals = np.linspace(chi_max, chi_min, 10000)
phi1, r1 = orbita_primer_tipo(chi_vals)

# ----- Órbita de segundo tipo (parte del círculo hacia la singularidad) -----
# ξ comienza en 0 (círculo) y va hacia π (singularidad)
xi_vals = np.linspace(0.01, np.pi - 0.01, 10000)
phi2, r2 = orbita_segundo_tipo(xi_vals)

# Ajustar los ángulos φ para una visualización continua
phi1 = phi1 - phi1[0]  # comenzar en φ=0
phi2 = phi2 - phi2[0]  # comenzar en φ=0

# Para conectar visualmente las órbitas, podemos rotar la segunda
phi2 = phi2 + phi1[-1]  # conectar al final de la primera órbita

# ======================
# COORDENADAS CARTESIANAS
# ======================
x1 = r1 * np.cos(phi1)
y1 = r1 * np.sin(phi1)

x2 = r2 * np.cos(phi2)
y2 = r2 * np.sin(phi2)

# ======================
# GRÁFICA
# ======================
plt.figure(figsize=(10, 6))

# Trazar las órbitas
plt.plot(x1, y1, 'b-', linewidth=2, label='Órbita de primer tipo')
plt.plot(x2, y2, 'r-', linewidth=2, label='Órbita de segundo tipo ')


# Círculos importantes

circle_critico = plt.Circle((0, 0), r_critico, color='black', linestyle=':', 
                           fill=False, linewidth=2, label=f'Círculo crítico r={r_critico:.2f}M')



plt.gca().add_patch(circle_critico)


# Configuración de la gráfica
plt.xlabel('x/M', fontsize=12)
plt.ylabel('y/M', fontsize=12)
plt.title('Órbitas de primer y segundo tipo con $2\mu(3+e)=1$', fontsize=14, pad=20)
plt.axis('equal')
plt.legend(loc='upper right', fontsize=10)
plt.grid(True, alpha=0.3)
plt.plot(0, 0, 'ko', markersize=7)

# Límites apropiados
max_radius = max(np.max(r1), np.max(r2), 6*M) * 1.2
plt.xlim(-max_radius, max_radius)
plt.ylim(-max_radius, max_radius)

plt.tight_layout()
plt.show()

# ======================
# GRÁFICA ADICIONAL: Radio vs Ángulo
# ======================
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 6))

# Radio vs φ para primera órbita
ax1.plot(phi1, r1, 'b-', linewidth=2)
ax1.axhline(y=2*M, color='black', linestyle='--', label='r = 2M (horizonte)')
ax1.set_xlabel(r'$\phi$ (rad)', fontsize=12)
ax1.set_ylabel('r/M', fontsize=12)
ax1.set_title('Órbita de primer tipo: Radio vs Ángulo')
ax1.legend()
ax1.grid(True, alpha=0.3)

# Radio vs φ para segunda órbita
ax2.plot(phi2, r2, 'r-', linewidth=2)

ax2.axhline(y=2*M, color='black', linestyle='--', label='r = 2M (horizonte)')
ax2.set_xlabel(r'$\phi$ (rad)', fontsize=12)
ax2.set_ylabel('r/M', fontsize=12)
ax2.set_title('Órbita de segundo tipo: Radio vs Ángulo')
ax2.legend()
ax2.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()

print(f"\nResumen de las órbitas:")
print(f"Órbita de primer tipo: Comienza en r ≈ {r1[0]:.2f}M, termina en r ≈ {r1[-1]:.2f}M")
print(f"Órbita de segundo tipo: Comienza en r ≈ {r2[0]:.2f}M, termina en r ≈ {r2[-1]:.2f}M")
print(f"Radio crítico: r = {r_critico:.2f}M")
```

### Orbitas ligadas con excentricidad imaginaria.

Considerando el caso $f(u)=0$ permite sólo una raíz real, la cual es necesariamente positiva para orbitas ligadas, y un par de raíces complejas conjugadas. Es evidente para las correspondientes órbitas que, comenzando en un afelio finito, caerá en la singularidad, circulando el origen una o más veces antes de caer.

Considerando orbitas con \textit{excentricidades imaginarias} (e \leq 0) y escribir las raíces de $f(u)=0$ como


$$
u_1=\frac{1}{2M}-\frac{2}{l}, \quad u_2=\frac{1}{l}(1+ie), \quad u_3=\frac{1}{l}(1-ie)
$$

Bajo las suposicion anterior, se considera:

$$\left(\frac{du}{d \phi}\right)^2 = 2M \left( u - \frac{1}{2M}+\frac{2}{l}\right) \left[ \left(u- \frac{1}{l}\right) + \frac{e^2}{l^2}\right] $$


Ahora, haciendo la sustitución

$$u = \frac{1}{l}(1+e\tan \frac{1}{2}\xi), \quad \ \frac{1}{2M}- \frac{2}{l} \leq u \leq \infty, \quad \ \xi_0 \leq \xi \leq \pi $$

donde 

$$ \tan \frac{1}{2}\xi_0 = - \frac{6\mu - 1}{2\mu e}, \quad \sin \frac{1}{2}\xi_0 = -\frac{6\mu - 1}{\Delta}, \quad \cos \frac{1}{2}\xi_0 = \frac{2\mu e}{\Delta} $$

$$\Delta = \left[(6\mu - 1)^2 + 4\mu^2 e^2 \right]$$

ahora,

$$\frac{d \xi}{d \phi} = \pm \sqrt{2}\left[(6\mu - 1) + 2\mu e \sin \xi + (6\mu -1)\cos \xi \right]^{1/2} $$

Por teoría de integrales elípticas, la solución de $\phi$ puede ser expresada en terminos de la \textit{integral elíptica Jacobiana}

$$\pm \phi = \frac{1}{\sqrt{\Delta}}\int^\phi \frac{d \gamma}{\sqrt{1-k^2 \sin^2 \gamma}}$$

$$k^2 = \frac{1}{2\Delta}(\Delta + 6\mu -1) $$

y

$$ \sin^2 \phi = \frac{1}{\Delta + 6\mu -1} \left[\Delta - 2\mu e \sin \xi - (6\mu - 1)\cos \xi \right] $$
$$= \frac{1}{\Delta + 6\mu -1} \left[\Delta + 6\mu -1 -2(2\mu e \sin \frac{1}{2}\xi + (6\mu - 1)\cos \frac{1}{2}\xi )\cos \frac{1}{2}\xi \right] $$

$$\left\{\begin{matrix}
\sin \psi =1&  \text{cuando}& \xi=\xi_0 & (\text{en el afelio}) \\
 & y &  \xi = \pi & (\text{en la singularidad})  \\
\end{matrix}\right.$$


también,

$$ \sin^2 \psi = 0 \quad \text{cuando} \quad \ \xi=\tan^{-1} \frac{2\mu e}{6\mu - 1}$$

De esta forma, $\psi$ asume que, el valor cero entre el rango $\xi_0 \leq \xi \leq \pi$ de $\xi$. Se concluye que el rango de $psi$ asociado con el rango de $\xi$ es

$$-\pi/2 \leq \psi \leq \pi/2 \quad \quad (\xi_0 \leq \xi \leq pi) $$

de esta forma podemos escribir la solución para $\phi$ como

$$\phi = \frac{1}{\sqrt{\Delta} \left[K(k), F(\psi, k) \right]}$$

donde $K(k)$ corresponde a la integral elíptica completa y $F(\psi, k)$ es la incompleta. Se asume que  $\phi=0$ en la singularidad donde $\xi=\pi$ y $\psi=\pi/2$, mientras el valor de $\phi$ en el afelio, donde $\xi=\xi_0$ y $\psi = - \frac{\pi}{2}$ es 

$$\phi_{ap}= 2K(k)/ \sqrt{\Delta} $$



```python
import numpy as np
import matplotlib.pyplot as plt

# Parámetros como en el PDF (Fig. 7a g y h)
M = 3/14
l = 1.0  # latus rectum
epsilon_values = [0.01, 0.1]  # valores de epsilon para e = i*epsilon

# Configuración de gráficos
fig, axes = plt.subplots(1, 2, figsize=(14, 6))

for idx, epsilon in enumerate(epsilon_values):
    print(f"\n=== Procesando epsilon = {epsilon} ===")
    
    # Parámetros derivados
    mu = M / l
    print(f"mu = M/l = {mu:.6f}")
    
    # Calcular constantes de movimiento usando ecuaciones (172)
    # 1/L^2 = (1/(lM)) * [1 - mu*(3 - epsilon^2)]
    L_sq_inv = (1/(l*M)) * (1 - mu*(3 - epsilon**2))
    L_sq = 1/L_sq_inv if abs(L_sq_inv) > 1e-12 else 1e12
    L = np.sqrt(L_sq)
    print(f"L = {L:.6f} M")
    
    # (1-E^2)/L^2 = (1/l^2) * (1-4*mu)*(1+epsilon^2)
    E_sq_term = L_sq * (1/(l**2)) * (1 - 4*mu) * (1 + epsilon**2)
    E_sq = 1 - E_sq_term
    print(f"E^2 = {E_sq:.6f}")
    
    # Ecuación diferencial: (du/dϕ)^2 = f(u)
    def f_u(u):
        return 2*M*u**3 - u**2 + (2*M/L_sq)*u - (1-E_sq)/L_sq
    
    # Encontrar el radio de afelio resolviendo f(u)=0
    # Buscamos la raíz real positiva más grande
    u_min = 0.01
    u_max = 10.0
    u_test = np.linspace(u_min, u_max, 10000)
    f_test = f_u(u_test)
    
    # Encontrar cambios de signo
    sign_changes = []
    for i in range(len(u_test)-1):
        if f_test[i] == 0:
            sign_changes.append(u_test[i])
        elif f_test[i] * f_test[i+1] < 0:
            # Interpolación lineal para raíz más precisa
            u_root = u_test[i] - f_test[i] * (u_test[i+1] - u_test[i]) / (f_test[i+1] - f_test[i])
            sign_changes.append(u_root)
    
    # Filtrar raíces positivas
    positive_roots = [r for r in sign_changes if r > 0]
    
    if positive_roots:
        u_ap = max(positive_roots)  # El más grande es el afelio
        r_ap = 1/u_ap
        print(f"Radio de afelio encontrado: u_ap = {u_ap:.6f}, r_ap = {r_ap:.6f} M")
    else:
        # Valor por defecto si no se encuentra raíz
        u_ap = 0.5
        r_ap = 1/u_ap
        print(f"Usando valor por defecto: u_ap = {u_ap:.6f}, r_ap = {r_ap:.6f} M")
    
    # Verificar que f(u_ap) ≈ 0
    print(f"f(u_ap) = {f_u(u_ap):.6e}")
    
    # Método de integración directo: integrar dϕ/du = -1/sqrt(f(u))
    # desde u_ap (afelio) hasta u grande (singularidad)
    
    # Número de puntos
    n_points = 20000
    
    # Crear array de u desde u_ap hasta u_max (singularidad)
    # Usar escala logarítmica para mejor resolución cerca de la singularidad
    u_start = u_ap
    u_end = 1000  # Corresponde a r = 0.001 M
    
    # Puntos espaciados logarítmicamente
    u_vals = np.logspace(np.log10(u_start), np.log10(u_end), n_points)
    
    # Inicializar arrays
    phi_vals = [0.0]  # ϕ comienza en 0 en el afelio
    r_vals = [1/u_start]
    
    # Integrar usando método trapezoidal
    for i in range(1, len(u_vals)):
        u_prev = u_vals[i-1]
        u_curr = u_vals[i]
        
        # f(u) en los puntos
        f_prev = f_u(u_prev)
        f_curr = f_u(u_curr)
        
        # Verificar que f(u) sea positiva
        if f_prev <= 0 or f_curr <= 0:
            print(f"Advertencia: f(u) no positiva en u={u_prev:.4f} o u={u_curr:.4f}")
            break
        
        # dϕ/du = -1/sqrt(f(u)) (negativo porque ϕ aumenta cuando u aumenta)
        dphi_du_prev = -1/np.sqrt(f_prev)
        dphi_du_curr = -1/np.sqrt(f_curr)
        
        # Integración trapezoidal: Δϕ = (dϕ/du_prev + dϕ/du_curr)/2 * Δu
        du = u_curr - u_prev
        delta_phi = 0.5 * (dphi_du_prev + dphi_du_curr) * du
        
        phi_vals.append(phi_vals[-1] + delta_phi)
        r_vals.append(1/u_curr)
    
    # Convertir a arrays numpy
    phi_vals = np.array(phi_vals)
    r_vals = np.array(r_vals)
    
    print(f"Integración completada: {len(phi_vals)} puntos")
    print(f"ϕ: {phi_vals[0]:.4f} → {phi_vals[-1]:.4f} rad")
    print(f"r: {r_vals[0]:.4f} → {r_vals[-1]:.4f} M")
    
    # Convertir a coordenadas cartesianas (dividiendo por M)
    x_vals = r_vals * np.cos(phi_vals) / M
    y_vals = r_vals * np.sin(phi_vals) / M
    
    # Graficar
    ax = axes[idx]
    
    # Trazar la órbita completa (desde afelio hasta singularidad)
    ax.plot(x_vals, y_vals, 'b-', linewidth=2, label=f'Órbita (e={epsilon})')
    
    # Marcar puntos importantes
    ax.plot(0, 0, 'ko', markersize=10, label='Singularidad (r=0)')
    
    ax.plot(x_vals[0], y_vals[0], 'go', markersize=8, label='Afelio')
    
    # Para ver mejor la estructura, también podemos trazar la parte simétrica
    # (si la órbita es simétrica respecto al afelio)
    #if idx == 0:  # Solo para el primer gráfico
        # Reflejar la órbita
        #x_vals_sym = np.concatenate([x_vals[::-1], x_vals])
        #y_vals_sym = np.concatenate([-y_vals[::-1], y_vals])
        #ax.plot(x_vals_sym, y_vals_sym, 'b--', linewidth=1, alpha=0.7, label='Órbita completa')
    
    # Configurar el gráfico
    ax.set_aspect('equal', adjustable='datalim')
    ax.set_title(f'Órbita con $e = {epsilon}i$', fontsize=14)
    ax.set_xlabel('$x/M$', fontsize=12)
    ax.set_ylabel('$y/M$', fontsize=12)
    ax.grid(True, alpha=0.3, linestyle='--')
    ax.axhline(0, color='black', linewidth=0.5, alpha=0.5)
    ax.axvline(0, color='black', linewidth=0.5, alpha=0.5)
    
    
    
    # Establecer límites apropiados para los ejes
    max_range = max(1.2 * max(abs(x_vals)), 1.2 * max(abs(y_vals)), 5)
    ax.set_xlim(-max_range, max_range)
    ax.set_ylim(-max_range, max_range)
    
    # Solo mostrar leyenda para puntos importantes
    ax.legend(loc='upper right', fontsize=9)



# Ajustar layout
plt.suptitle('Geodésicas tipo-tiempo ligadas con excentricidades imaginarias$ ', fontsize=16, y=1.02)
plt.tight_layout()
plt.show()

# Mostrar información de depuración para verificar cálculos
print("\n" + "="*70)
print("RESUMEN DE CÁLCULOS PARA VERIFICACIÓN:")
print("="*70)

for epsilon in epsilon_values:
    mu = M / l
    L_sq_inv = (1/(l*M)) * (1 - mu*(3 - epsilon**2))
    L_sq = 1/L_sq_inv if abs(L_sq_inv) > 1e-12 else 1e12
    E_sq_term = L_sq * (1/(l**2)) * (1 - 4*mu) * (1 + epsilon**2)
    E_sq = 1 - E_sq_term
    
    print(f"\nPara e = {epsilon}:")
    print(f"  μ = M/l = {mu:.8f}")
    print(f"  1 - μ(3 - ε²) = {1 - mu*(3 - epsilon**2):.8f}")
    print(f"  L² = {L_sq:.8f} M²")
    print(f"  (1-4μ)(1+ε²) = {(1-4*mu)*(1+epsilon**2):.8f}")
    print(f"  E² = {E_sq:.8f}")
    print(f"  1-E² = {1-E_sq:.8f}")
```

## (c) Las órbitas no ligadas (E^2 > 1)

### (i) Órbitas de primera y segunda clase

Cuando todas las raíces son reales, se continua expresando en terminos de una excentricidad $e$ como en el caso de órbitas ligadas, con la diferencia que ahora $e\geq 1$, por lo que las raíces son escritas como

\begin{equation}
  u_1=-\frac{1}{l}(e-1), \quad u_2= \frac{1}{l}(e+1), \quad u_3=\frac{1}{2M}-\frac{2}{l}
\end{equation}

\begin{equation}
\frac{1}{L^2}=\frac{1}{l M} [1-\mu (3+e^2)], \quad \ \frac{E^2-1}{L^2}=\frac{1}{l^2}(1-4\mu)(e^2 - 1)
\end{equation}

Dado $L^2 > 0$ y $E^2-1 \geq 0 $ (asumido),

\begin{equation}
1- \mu(3 + e^2), \quad \text{y} \quad \mu\leq \frac{1}{4}
\end{equation}

De esta forma, cuando las raíces $u_2=u_3$, la relacion (192) se transforma

\begin{equation}
\frac{L^2}{M^2}=4 \frac{(3+e)^2}{(3-e)(e+1)}, \quad E^2 - 1=\frac{e^2 - 1}{9 - e^2}
\end{equation}

para estas orbitas en especial, el rango para $e$:

\begin{equation}
 1 \leq e \leq 3
\end{equation}

y los correspondientes distancias a perihelios deben estar en este rango, (como en el caso de las orbitas ligadas)

\begin{equation}
3M < r_p \leq 4M 
\end{equation}

Se nota, que dado que las órbitas de primera clase (y las órbitas pertenecientes a excentricidad imaginaria) "llegan desde el infinito", así, se asocia con un \textit{parámetro de impacto} $D$, una velocidad en el infinito $v_{\infty}$; relacionadas a $L$ y $E$ por


\begin{equation}
D^2= \frac{L^2}{v_{\infty}^2}= \frac{L^2 E^2}{E^2-1} 
\end{equation}

con $E^2 = \frac{1}{1-v_{\infty}^2}$. Ahora, haciendo la sustitución:

\begin{equation}
u=\frac{1}{l}(1+e \cos \xi)
\end{equation}

Sin embargo, considerando que $e\geq 1$,

\begin{equation}
u=0 \quad \ \text{cuando } \quad \chi = \cos^{-1}(-e^{-1}=\chi_{\infty})
\end{equation}

Pero en el perihelio sigue ocurriendo cuando $\chi=0$. El rango permitido para $\chi$, es

\begin{equation}
0 \geq \chi < \chi_{\infty} = \cos^{-1}(-e^{-1}) 
\end{equation}

Considerando las restricciones para $\chi$, podemos escribir la solución para $\phi$.

\begin{equation}
\phi = \frac{2}{(1-6\mu +2\mu e)^{1/2}} \left[K(k)- F(\pi/2 - \chi/2 , k) \right]
\end{equation}

Se ha tomado el origen de $\phi$ en el perihelio donde $\chi=0$. La trayectoria se va a infinito, asintóticamente, en la dirección:

\begin{equation}
\phi = \phi_{\infty} = \frac{2}{(1-6\mu +2\mu e)^{1/2}}  \left[K(k) - F(\psi_{\infty},k) \right]
\end{equation}

donde $\psi_{\infty}=\frac{1}{2}\cos^{-1}e^{-1}$. Las órbitas no ligadas se acercan al círculo en $r_p$, asintóticamente, girando en espiral alrededor de él un número infinito de veces.

### (ii) Las órbitas con excentricidad imaginaria

Las órbitas no ligadas, caen en la singularidad, caracterizadas por una excentricidad imaginaria, con el cambio que en lugar de partir de un afelio finito, llegan desde el infinito. 

\begin{equation}
\frac{1}{L^2}=\frac{1}{lM}[1- \mu(3-e^2)] \quad \ \ \frac{E^2-1}{L^2}=\frac{1}{l^2}(4\mu - 1)(1+e^2)
\end{equation}

se requiere,

\begin{equation}
\mu \geq \frac{1}{4}, \quad 1-3\mu+\mu e^2  >0
\end{equation}

Para $\frac{1}{3}\geq \mu \geq \frac{1}{4}$, la segunda inigualdad no impone restricciones en $e^2$, pero para $\mu  >\frac{1}{3}$; pero para $\mu > \frac{1}{3}$ es necesario imponer la restriccion:

\begin{equation}
e^2  > 3 - \frac{1}{\mu}
\end{equation}

Para obtener una solución a aquellas órbitas, se hace la sustitución 

\begin{equation}
u= \frac{1}{l}(1+e\tan\frac{1}{2}\xi)
\end{equation}

Sin embargo, en vista de la presente restricción ($\mu \geq \frac{1}{4}$), el rango de $\xi$ debe terminar en $\xi_{\infty}$ (antes  $\xi_0$), donde

\begin{equation}
\begin{aligned}
\tan \frac{1}{2}\xi_{\infty}
&=-e^{-1}\\
\sin \frac{1}{2}\xi_{\infty} 
=& -\frac{1}{\sqrt{1+e^2}} \\
\cos \frac{1}{2}\xi_{\infty} = \frac{e}{\sqrt{1+e^2}} \\
\end{aligned}
\end{equation}

Cuando $\xi = \xi_\infty$, $u$ se hace cero y cuando $\xi < \xi_{\infty}$, $u$ se vuelve negativa. Por lo tanto, como $\xi \rightarrow \xi_\infty$ acercandose por la derecha de cero, $u \rightarrow 0$ entonces $r \rightarrow \infty$. Pero el límite superior en $\xi$, sea $\pi$ (cuando $u\rightarrow \infty $ y $r\rightarrow 0$), no es afectado. El rango de $\xi$:

\begin{equation}
\xi_{\infty} < \xi \leq \pi
\end{equation}

A partir de estos cambios, la solución para $\phi$ puede ser escrita como

\begin{equation}
\phi=\frac{1}{\sqrt{\Delta}} \left[K(k)-F(\psi,k) \right]
\end{equation}

El origen de $\phi$ (como en la solución anterior) es en la singularidad (donde $\xi = \pi$ y $\psi=\frac{\pi}{2}$). Pero el límite inferior en $\psi$, sea $\psi_\infty$, mientras que la órbita sigue hasta el infinito, puede ser obtenida de $$ \sin^2 \phi = \frac{1}{\Delta + 6\mu -1} \left[\Delta - 2\mu e \sin \xi - (6\mu - 1)\cos \xi \right] $$
$$= \frac{1}{\Delta + 6\mu -1} \left[\Delta + 6\mu -1 -2(2\mu e \sin \frac{1}{2}\xi + (6\mu - 1)\cos \frac{1}{2}\xi )\cos \frac{1}{2}\xi \right] $$
al insertar $\sin \frac{1}{2}\xi$ y $\cos \frac{1}{2}\xi$ los valores apropiados para $\xi_{\infty}$ dados anteriormente (eq ?)

\begin{equation}
\sin^2 \psi_{\infty}=\frac{1}{\Delta + 6\mu - 1} (\Delta + 6\mu -1 - 2e^2\frac{4\mu - 1}{e^2 +1})
\end{equation}



```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.special import ellipk
import warnings
warnings.filterwarnings('ignore')

# Parámetros dados
M = 3/14
e = 1.5  # excentricidad
l_vals = [4.5, 2.5, 1.94]  # latera recta para cada caso

def orbita_primera_clase(l, chi_array):
    """
    Calcula r(phi) para una órbita de primera clase (no ligada)
    con parámetros l y e. chi va de -chi_inf a chi_inf, con chi=0 en perihelio.
    """
    mu = M / l
    k2 = (4 * mu * e) / (1 - 6 * mu + 2 * mu * e)
    

    if k2 <= 0 or k2 >= 1:
        k2 = 0.5 
    
    # integral elíptica completa K(k)
    Kk = ellipk(k2)
    
    # phi(chi)
    phi_vals = []
    for chi in chi_array:
        psi = np.pi/2 - chi/2
        # Integral elíptica incompleta F(ψ, k)
        #integración numérica
        if psi > 0:
            theta_vals = np.linspace(0, psi, 200)
            integrand = 1 / np.sqrt(1 - k2 * np.sin(theta_vals)**2)
            F = np.trapz(integrand, theta_vals)
        else:
            F = 0
        
        phi = (2 / np.sqrt(abs(1 - 6*mu + 2*mu*e))) * (Kk - F)
        phi_vals.append(phi)
    
    phi_vals = np.array(phi_vals)
    
    # r(chi)
    r_vals = l / (1 + e * np.cos(chi_array))
    
    # Coordenadas polares a cartesianas
    x_vals = r_vals * np.cos(phi_vals)
    y_vals = r_vals * np.sin(phi_vals)
    
    return x_vals, y_vals, phi_vals, r_vals

def orbita_segunda_clase(l, num_points=800):
    """
    Calcula la órbita de segunda clase usando la formulación analítica
    basada en la sustitución (154) y ecuación (157) del texto.
    """
    mu = M / l
    
    # k² según la ecuación (128)
    k2 = (4 * mu * e) / (1 - 6 * mu + 2 * mu * e)
    
    # Aseguramos que k² sea válido
    if k2 <= 0 or k2 >= 1:
        k2 = 0.5
    
    # Discretizamos ξ de 0 a π (ξ=0 en aphelion, ξ=π en singularidad)
    xi_vals = np.linspace(0.01, np.pi-0.01, num_points)
    
    # Calculamos u según (154)
    u3 = 1/(2*M) - 2/l
    coeff = 1/(2*M) - (3+e)/l
    
    # Verificamos que el coeficiente no sea negativo
    coeff = abs(coeff)
    
    u_vals = u3 + coeff * np.tan(xi_vals/2)**2
    r_vals = 1/u_vals
    
    # Calculamos φ usando (157)
    phi_vals = []
    Kk = ellipk(k2)
    
    for xi in xi_vals:
        # Integral elíptica incompleta F(ξ/2, k)
        theta_vals = np.linspace(0, xi/2, 200)
        integrand = 1 / np.sqrt(1 - k2 * np.sin(theta_vals)**2)
        F_val = np.trapz(integrand, theta_vals)
        phi = (2 / np.sqrt(abs(1 - 6*mu + 2*mu*e))) * F_val
        phi_vals.append(phi)
    
    phi_vals = np.array(phi_vals)
    
    # Coordenadas cartesianas
    x_vals = r_vals * np.cos(phi_vals)
    y_vals = r_vals * np.sin(phi_vals)
    
    return x_vals, y_vals, phi_vals, r_vals

# ============================================================================
# FIGURA 1: Órbitas (primera y segunda clase)
# ============================================================================

print("=" * 60)
print("FIGURA 1: Órbitas completas (primera y segunda clase)")
print("=" * 60)

fig1, axes1 = plt.subplots(1, 3, figsize=(18, 6), subplot_kw={'aspect': 'equal'})

for idx, l in enumerate(l_vals):
    ax = axes1[idx]
    mu = M / l
    
    # Información sobre las raíces
    u1 = -1/l * (e - 1)
    u2 = 1/l * (e + 1)
    u3 = 1/(2*M) - 2/l
    
    print(f"\nl={l}:")
    print(f"  μ = M/l = {mu:.4f}")
    print(f"  Perihelion: r_p = {1/u2:.4f}M")
    print(f"  Aphelion (2nd kind): r_ap = {1/u3:.4f}M")
    
    # ========================================================================
    # ÓRBITA DE PRIMERA CLASE 
    # ========================================================================
    
    # Calcular chi_inf (donde r → ∞)
    chi_inf = np.arccos(-1/e)
    
    # Crear array de chi que va desde -chi_inf hasta chi_inf
    # chi=0 en perihelio, chi negativo para la trayectoria de llegada,
    # chi positivo para la trayectoria de salida
    chi_arrival = np.linspace(-chi_inf + 1e-3, 0, 400)  # Llegada al perihelio
    chi_departure = np.linspace(0, chi_inf - 1e-3, 400)  # Salida desde perihelio
    chi_full = np.concatenate([chi_arrival, chi_departure[1:]])  # Excluimos el 0 duplicado
    
    # Calcular la órbita completa
    x1_full, y1_full, phi1_full, r1_full = orbita_primera_clase(l, chi_full)
    
    # Separar en llegada y salida
    arrival_idx = len(chi_arrival)
    x1_arrival = x1_full[:arrival_idx]
    y1_arrival = y1_full[:arrival_idx]
    x1_departure = x1_full[arrival_idx-1:]  # Incluimos el perihelio
    y1_departure = y1_full[arrival_idx-1:]  # Incluimos el perihelio

    # AGREGAR FLECHAS DE DIRECCIÓN
    # ========================================================================
    perihelion_idx= len(chi_arrival) - 1
    # Flecha en la trayectoria de llegada
    arrival_arrow_idx = perihelion_idx // 3
    if arrival_arrow_idx > 1:
        dx_arr = x1_full[arrival_arrow_idx] - x1_full[arrival_arrow_idx-1]
        dy_arr = y1_full[arrival_arrow_idx] - y1_full[arrival_arrow_idx-1]
        ax.arrow(x1_full[arrival_arrow_idx], y1_full[arrival_arrow_idx], 
                dx_arr, dy_arr,
                head_width=0.2, head_length=0.3, 
                fc='darkblue', ec='darkblue', alpha=0.8)
    
    # Flecha en la trayectoria de salida
    departure_arrow_idx = perihelion_idx + (len(x1_full) - perihelion_idx) // 3
    if departure_arrow_idx < len(x1_full) - 1:
        dx_dep = x1_full[departure_arrow_idx+1] - x1_full[departure_arrow_idx]
        dy_dep = y1_full[departure_arrow_idx+1] - y1_full[departure_arrow_idx]
        ax.arrow(x1_full[departure_arrow_idx], y1_full[departure_arrow_idx], 
                dx_dep, dy_dep,
                head_width=0.2, head_length=0.3, 
                fc='darkblue', ec='darkblue', alpha=0.8)
    



    
    









    # Graficar órbita de primera clase
    ax.plot(x1_arrival, y1_arrival, 'b-', linewidth=2.5, alpha=0.7, label='Primera clase')
    ax.plot(x1_departure, y1_departure, 'b-', linewidth=2.5, alpha=0.7, )
    
    # Marcar el perihelio
    #perihelion_idx = arrival_idx - 1
    #ax.plot(x1_full[perihelion_idx], y1_full[perihelion_idx], 'ro', 
            #markersize=10, markeredgecolor='k', linewidth=1.5, label='Perihelion')
    
    # ========================================================================
    # ÓRBITA DE SEGUNDA CLASE
    # ========================================================================
    
    x2, y2, phi2, r2 = orbita_segunda_clase(l, num_points=600)
    ax.plot(x2, y2, 'r-', linewidth=2.5, alpha=0.7, label='Segunda clase')
    
    # Marcar el afelio de la órbita de segunda clase
    #ax.plot(x2[0], y2[0], 'rs', markersize=8, markeredgecolor='k', 
            #linewidth=1.5, label='Aphelion (2nd)')
    
    # ========================================================================
    # ELEMENTOS ADICIONALES DEL GRÁFICO
    # ========================================================================
    
    # Círculos importantes
    theta = np.linspace(0, 2*np.pi, 300)
    
    # Horizonte (r = 2M)
    horizon_radius = 2*M
    ax.plot(horizon_radius*np.cos(theta), horizon_radius*np.sin(theta), 
            'k--', alpha=0.8, linewidth=2, label='Horizon (2M)')
    
    # Órbita circular crítica (r = 3M)
    #critical_radius = 3*M
    #ax.plot(critical_radius*np.cos(theta), critical_radius*np.sin(theta), 
            #'g-.', alpha=0.8, linewidth=2, label='Critical (3M)')
    
    # Círculo de perihelio
    #r_peri = l/(1+e)
    #if r_peri > 0:
        #ax.plot(r_peri*np.cos(theta), r_peri*np.sin(theta), 
                #'m:', alpha=0.5, linewidth=1.5, label=f'Perihelion circle')
    
    # ========================================================================
    # CONFIGURACIÓN DEL GRÁFICO
    # ========================================================================
    
    ax.set_xlabel('x/M', fontsize=12)
    ax.set_ylabel('y/M', fontsize=12)
    ax.set_title(f'Case {idx+1}: l = {l:.2f}, μ = {mu:.3f}', fontsize=14, fontweight='bold')
    ax.plot(0, 0, 'ko', markersize=8, label='Singularidad')
    ax.grid(True, linestyle='--', alpha=0.3)
    
    # Configurar leyenda
    if idx == 0:
        ax.legend(loc='upper left', fontsize=9, framealpha=0.9)
    else:
        ax.legend(loc='upper right', fontsize=9, framealpha=0.9)
    
    # Configurar límites del gráfico
    max_range = 3
    ax.set_xlim(-max_range, max_range)
    ax.set_ylim(-max_range, max_range)
    
    # Agregar ejes
    ax.axhline(y=0, color='k', linestyle='-', alpha=0.2, linewidth=0.5)
    ax.axvline(x=0, color='k', linestyle='-', alpha=0.2, linewidth=0.5)

plt.suptitle('Geodésicas tipo-tiempo con $E^2 > 1$, $e=3/2$\nPrimera clase (azul) y Segunda clase (rojo)', 
             fontsize=16, fontweight='bold', y=1.02)
plt.tight_layout()
plt.show()

# ============================================================================
# FIGURA 2: Solo órbitas de primera clase con direcciones asintóticas
# ============================================================================

print("\n" + "=" * 60)
print("FIGURA 2: Solo órbitas de primera clase")
print("=" * 60)

fig2, axes2 = plt.subplots(1, 3, figsize=(18, 6), subplot_kw={'aspect': 'equal'})

for idx, l in enumerate(l_vals):
    ax = axes2[idx]
    mu = M / l
    
    # Calcular chi_inf
    chi_inf = np.arccos(-1/e)
    
    # Crear array de chi completo (llegada + salida)
    chi_arrival = np.linspace(-chi_inf + 1e-3, 0, 400)
    chi_departure = np.linspace(0, chi_inf - 1e-3, 400)
    chi_full = np.concatenate([chi_arrival, chi_departure[1:]])
    
    # Calcular órbita
    x1, y1, phi1, r1 = orbita_primera_clase(l, chi_full)
    
    # Índice del perihelio
    perihelion_idx = len(chi_arrival) - 1
    
    # Graficar órbita completa
    ax.plot(x1, y1, 'b-', linewidth=3, alpha=0.8, label=f'Orbit (l={l})')
    
    # Marcar perihelio
    #ax.plot(x1[perihelion_idx], y1[perihelion_idx], 'ro', 
            #markersize=12, markeredgecolor='k', linewidth=2, 
            #label='Perihelion', zorder=10)
    
    # ========================================================================
    # AGREGAR FLECHAS DE DIRECCIÓN
    # ========================================================================
    
    # Flecha en la trayectoria de llegada
    arrival_arrow_idx = perihelion_idx // 3
    if arrival_arrow_idx > 1:
        dx_arr = x1[arrival_arrow_idx] - x1[arrival_arrow_idx-1]
        dy_arr = y1[arrival_arrow_idx] - y1[arrival_arrow_idx-1]
        ax.arrow(x1[arrival_arrow_idx], y1[arrival_arrow_idx], 
                dx_arr, dy_arr,
                head_width=0.2, head_length=0.3, 
                fc='darkblue', ec='darkblue', alpha=0.8)
    
    # Flecha en la trayectoria de salida
    departure_arrow_idx = perihelion_idx + (len(x1) - perihelion_idx) // 3
    if departure_arrow_idx < len(x1) - 1:
        dx_dep = x1[departure_arrow_idx+1] - x1[departure_arrow_idx]
        dy_dep = y1[departure_arrow_idx+1] - y1[departure_arrow_idx]
        ax.arrow(x1[departure_arrow_idx], y1[departure_arrow_idx], 
                dx_dep, dy_dep,
                head_width=0.2, head_length=0.3, 
                fc='darkblue', ec='darkblue', alpha=0.8)
    
    # ========================================================================
    # AGREGAR LÍNEAS ASINTÓTICAS
    # ========================================================================
    
    # Calcular ángulos asintóticos
    phi_arrival = phi1[0]  # Ángulo de llegada desde infinito
    phi_departure = phi1[-1]  # Ángulo de salida hacia infinito
    
    # Calcular deflexión total
    deflection = phi_departure - phi_arrival
    
    # Línea asintótica de llegada
    r_asym = 20
    x_in = r_asym * np.cos(phi_arrival)
    y_in = r_asym * np.sin(phi_arrival)
    ax.plot([x_in, x1[0]], [y_in, y1[0]], 'k:', alpha=0.6, linewidth=1.5, 
            label='Asymptotic')
    
    # Línea asintótica de salida
    x_out = r_asym * np.cos(phi_departure)
    y_out = r_asym * np.sin(phi_departure)
    ax.plot([x1[-1], x_out], [y1[-1], y_out], 'k:', alpha=0.6, linewidth=1.5)
    
    # ========================================================================
    # ELEMENTOS ADICIONALES
    # ========================================================================
    
    # Círculos importantes
    theta = np.linspace(0, 2*np.pi, 300)
    ax.plot(2*M*np.cos(theta), 2*M*np.sin(theta), 'k--', alpha=0.7, linewidth=1.5)
    ax.plot(3*M*np.cos(theta), 3*M*np.sin(theta), 'g-.', alpha=0.7, linewidth=1.5)
    
    # ========================================================================
    # CONFIGURACIÓN DEL GRÁFICO
    # ========================================================================
    
    ax.set_xlabel('x/M', fontsize=12)
    ax.set_ylabel('y/M', fontsize=12)
    ax.set_title(f'l = {l:.2f}, Δφ = {deflection:.2f} rad', fontsize=14, fontweight='bold')
    ax.grid(True, linestyle='--', alpha=0.3)
    ax.legend(loc='upper right', fontsize=9, framealpha=0.9)
    
    # Configurar límites
    max_range = 4
    ax.set_xlim(-max_range, max_range)
    ax.set_ylim(-max_range, max_range)
    
    # Ejes
    ax.axhline(y=0, color='k', linestyle='-', alpha=0.2, linewidth=0.5)
    ax.axvline(x=0, color='k', linestyle='-', alpha=0.2, linewidth=0.5)

plt.suptitle('FIGURA 2: Orbits of the first kind only\nShowing full trajectory from infinity to infinity', 
             fontsize=16, fontweight='bold', y=1.02)
plt.tight_layout()
plt.show()

# ============================================================================
# FIGURA 3: Vista detallada de un caso
# ============================================================================

print("\n" + "=" * 60)
print("FIGURA 3: Vista detallada (l = 4.5)")
print("=" * 60)

l_detailed = 4.5
mu_detailed = M / l_detailed

fig3, ax3 = plt.subplots(1, 1, figsize=(10, 10), subplot_kw={'aspect': 'equal'})

# Calcular chi_inf
chi_inf = np.arccos(-1/e)

# Crear array de chi completo
chi_arrival = np.linspace(-chi_inf + 1e-3, 0, 600)
chi_departure = np.linspace(0, chi_inf - 1e-3, 600)
chi_full = np.concatenate([chi_arrival, chi_departure[1:]])

# Calcular órbita de primera clase
x1, y1, phi1, r1 = orbita_primera_clase(l_detailed, chi_full)

# Índice del perihelio
perihelion_idx = len(chi_arrival) - 1

# Graficar órbita completa
ax3.plot(x1, y1, 'b-', linewidth=3.5, alpha=0.9, label=f'Órbita de primera clase (l={l_detailed})')

# Destacar trayectoria de llegada y salida
#ax3.plot(x1[:perihelion_idx+1], y1[:perihelion_idx+1], 'b-', linewidth=4, alpha=0.5, label='Arrival')
#ax3.plot(x1[perihelion_idx:], y1[perihelion_idx:], 'c-', linewidth=4, alpha=0.5, label='Departure')

# Marcar perihelio
#ax3.plot(x1[perihelion_idx], y1[perihelion_idx], 'ro', 
        #markersize=15, markeredgecolor='k', linewidth=2.5, 
        #label='Perihelion', zorder=10)

# Calcular órbita de segunda clase
x2, y2, phi2, r2 = orbita_segunda_clase(l_detailed, num_points=800)
ax3.plot(x2, y2, 'r--', linewidth=3, alpha=0.8, label='Primera clase')

# Círculos importantes
theta = np.linspace(0, 2*np.pi, 500)

# Horizonte
ax3.plot(2*M*np.cos(theta), 2*M*np.sin(theta), 'k--', 
        linewidth=3, alpha=0.9, label='Horizon (2M)')

# Órbita crítica
#ax3.plot(3*M*np.cos(theta), 3*M*np.sin(theta), 'g-.', 
        #linewidth=3, alpha=0.9, label='Critical orbit (3M)')

# Círculo de perihelio
r_peri = l_detailed/(1+e)
#ax3.plot(r_peri*np.cos(theta), r_peri*np.sin(theta), 'm:', 
        #linewidth=2.5, alpha=0.7, label=f'Perihelion circle ({r_peri:.2f}M)')

# Ángulo de deflexión
phi_arrival = phi1[0]
phi_departure = phi1[-1]
deflection = phi_departure - phi_arrival

# Líneas asintóticas
r_asym = 25
x_in = r_asym * np.cos(phi_arrival)
y_in = r_asym * np.sin(phi_arrival)
x_out = r_asym * np.cos(phi_departure)
y_out = r_asym * np.sin(phi_departure)

ax3.plot([x_in, x1[0]], [y_in, y1[0]], 'k:', linewidth=2.5, alpha=0.7)
ax3.plot([x1[-1], x_out], [y1[-1], y_out], 'k:', linewidth=2.5, alpha=0.7)

# Flechas de dirección
# En llegada
arr_arrow_idx = perihelion_idx // 4
dx_arr = x1[arr_arrow_idx] - x1[arr_arrow_idx-1]
dy_arr = y1[arr_arrow_idx] - y1[arr_arrow_idx-1]
ax3.arrow(x1[arr_arrow_idx], y1[arr_arrow_idx], dx_arr, dy_arr,
         head_width=0.2, head_length=0.3, fc='darkblue', ec='darkblue', linewidth=2)

# En salida
dep_arrow_idx = perihelion_idx + (len(x1) - perihelion_idx) // 4
dx_dep = x1[dep_arrow_idx+1] - x1[dep_arrow_idx]
dy_dep = y1[dep_arrow_idx+1] - y1[dep_arrow_idx]
ax3.arrow(x1[dep_arrow_idx], y1[dep_arrow_idx], dx_dep, dy_dep,
         head_width=0.2, head_length=0.3, fc='darkblue', ec='darkblue', linewidth=2)

# Configuración del gráfico
ax3.set_xlabel('x/M', fontsize=14)
ax3.set_ylabel('y/M', fontsize=14)
ax3.set_title(f'Detailed view: e={e}, l={l_detailed}, μ={mu_detailed:.3f}\nDeflection angle Δφ = {deflection:.3f} rad', 
              fontsize=16, fontweight='bold', pad=20)

ax3.grid(True, linestyle='--', alpha=0.3)
ax3.legend(loc='upper left', fontsize=11, framealpha=0.95)

# Límites y ejes
max_range = 5
ax3.set_xlim(-max_range, max_range)
ax3.set_ylim(-max_range, max_range)

ax3.axhline(y=0, color='k', linestyle='-', alpha=0.3, linewidth=1)
ax3.axvline(x=0, color='k', linestyle='-', alpha=0.3, linewidth=1)

plt.tight_layout()
plt.show()

print("\nResumen de parámetros:")
print("-" * 40)
print(f"M = {M:.4f}")
print(f"e = {e}")
print(f"Horizon radius = {2*M:.4f} M")
print(f"Critical orbit radius = {3*M:.4f} M")
print(f"\nPara l = 4.5:")
print(f"  Perihelion radius = {l_detailed/(1+e):.4f} M")
print(f"  μ = M/l = {mu_detailed:.4f}")
print(f"  Deflection angle = {deflection:.4f} rad = {deflection*180/np.pi:.2f}°")
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.special import ellipk, ellipe
from scipy.integrate import quad, cumulative_trapezoid
import warnings
warnings.filterwarnings('ignore')

# ============================================================================
# PARÁMETROS SEGÚN FIGURA 7b (d) y (e)
# ============================================================================
M = 0.3                     # Masa en unidades de longitud
l = 1.0                     # Latus rectum
imag_es = [0.001, 0.1]     # Valores de e (parte imaginaria, sin la i)
colors = ['darkblue', 'darkred']
labels = ['e = 0.001i', 'e = 0.1i']

# ============================================================================
# FUNCIÓN PARA CALCULAR F(ψ, k) - INTEGRAL ELÍPTICA INCOMPLETA
# ============================================================================
def elliptic_f(psi, k):
    """
    Calcula la integral elíptica incompleta F(ψ, k)
    usando integración numérica precisa
    """
    if np.abs(psi) < 1e-12:
        return 0.0
    
    # Usamos quad para integración adaptativa
    def integrand(gamma):
        return 1.0 / np.sqrt(1.0 - k**2 * np.sin(gamma)**2)
    
    result, error = quad(integrand, 0, psi, epsabs=1e-12, epsrel=1e-12)
    return result

# ============================================================================
# FUNCIÓN PARA ÓRBITAS CON EXCENTRICIDAD IMAGINARIA (E² > 1)
# Usamos ecuaciones (204)-(210) y sustitución (175)
# ============================================================================
def orbita_excent_imaginaria(l, imag_e, M, num_points=800):
    """
    Calcula la órbita no ligada con excentricidad imaginaria (E² > 1).
    """
    mu = M / l
    e = imag_e  # e positivo (parte imaginaria)
    
    print(f"\n{'='*60}")
    print(f"Calculando órbita con e = {e}i, μ = {mu:.4f}, M = {M}")
    print(f"{'='*60}")
    
    # Para e imaginario: e² es negativo en algunas fórmulas
    # En las ecuaciones, cuando aparece e², es en realidad -(e)^2 para imaginario puro
    e2_neg = -(e**2)  # e² negativo
    
    # Según ecuaciones (204):
    # 1/L² = (1/(lM))[1 - μ(3 - e²)] pero para e imaginario, e² es negativo
    # 1 - μ(3 - e²) = 1 - μ(3 - (-e²)) = 1 - μ(3 + e²)
    
    # Δ según ecuación (180) pero con e imaginario
    # Δ = sqrt((6μ-1)² + 4μ²e²) con e² negativo
    Delta = np.sqrt((6*mu - 1)**2 + 4*mu**2 * e2_neg)
    print(f"Δ = sqrt((6μ-1)² + 4μ²e²) = {Delta:.6f}")
    
    # k² según ecuación (183)
    k2 = 0.5 * (Delta + 6*mu - 1) / Delta
    k = np.sqrt(np.abs(k2))
    print(f"k² = {k2:.6f}, k = {k:.6f}")
    
    # ξ_∞ según ecuaciones (207): tan(ξ_∞/2) = -1/e
    # Para e imaginario puro, esto es complejo. Debemos usar el límite apropiado.
    # En realidad, para e imaginario, el rango de ξ es diferente
    
    # Según el texto, para órbitas no ligadas con e imaginario:
    # ξ va desde ξ_∞ (donde u=0, r→∞) hasta π (donde u→∞, r=0)
    # ξ_∞ = 2*arctan(-1/e) pero esto da complejo para e real
    
    # Para e imaginario, reescribimos:
    # tan(ξ/2) = -1/(ie) = i/e (donde e es la magnitud)
    # Entonces ξ/2 = arctan(i/e) = i*arctanh(1/e) para e>1
    # Esto implica que ξ es imaginario puro
    
    # En lugar de esto complejo, usamos la formulación alternativa:
    # u = (1/l)(1 + e*tan(ξ/2)) con e imaginario
    
    # DISCRETIZACIÓN DIRECTA DE u
    # ========================================================================
    # rango de u: desde 0 (infinito) hasta ∞ (singularidad)
    # Usamos una discretización logarítmica
    
    # Valores de u desde muy pequeño hasta grande
    u_min = 1e-4  # r ≈ 10000
    u_max = 20.0  # r ≈ 0.05
    
    # Discretización logarítmica
    u_vals = np.logspace(np.log10(u_min), np.log10(u_max), num_points)
    r_vals = 1.0 / u_vals
    
    # Para cada u, necesitamos encontrar ξ
    # u = (1/l)(1 + e*tan(ξ/2)) ⇒ tan(ξ/2) = (l*u - 1)/e
    # Pero e es imaginario: e = i*|e|
    
    # Reescribimos: u = (1/l)(1 + i|e|*tan(ξ/2))
    # Entonces: tan(ξ/2) = -i*(l*u - 1)/|e|
    
    # Esto sugiere que ξ/2 es imaginario puro: tan(ξ/2) = i*y
    # donde y = -(l*u - 1)/|e|
    
    # Entonces: ξ = 2*i*arctanh(y) = i*ξ_img
    # donde ξ_img = 2*arctanh(y)
    
    # CALCULAR φ PARA CADA u
    # ========================================================================
    phi_vals = np.zeros_like(u_vals)
    Kk = ellipk(k2)  # Integral elíptica completa
    
    for i, u in enumerate(u_vals):
        # Calcular y = -(l*u - 1)/e
        y = -(l*u - 1) / e
        
        # ξ_img = 2*arctanh(y) (pero cuidado con dominio)
        if np.abs(y) < 1 - 1e-10:
            xi_img = 2.0 * np.arctanh(y)
        else:
            xi_img = np.sign(y) * 20.0  # Valor grande para el límite
            
        # ξ es imaginario puro: ξ = i*xi_img
        # Necesitamos ψ según ecuación (184)
        
        # Para ξ imaginario: sin(ξ) = i*sinh(xi_img), cos(ξ) = cosh(xi_img)
        sin_xi = 1j * np.sinh(xi_img)
        cos_xi = np.cosh(xi_img)
        
        # Calcular ψ según ecuación (184)
        numerator = Delta - 2*mu*e*sin_xi - (6*mu - 1)*cos_xi
        denominator = Delta + 6*mu - 1
        
        sin2_psi = numerator / denominator
        
        # ψ puede ser complejo, tomamos la parte real apropiada
        sin2_psi_real = np.real(sin2_psi)
        sin2_psi_real = np.clip(sin2_psi_real, 0, 1)
        
        psi = np.arcsin(np.sqrt(sin2_psi_real))
        
        # Asegurar que ψ esté en el cuadrante correcto
        # Basado en el texto, ψ va desde -π/2 hasta π/2
        if u < 1/l:  # r > l
            psi = -psi  # Para la rama de llegada
            
        # Calcular F(ψ, k)
        F_psi = elliptic_f(psi, k)
        
        # Calcular φ según ecuación (209)
        phi = (1.0 / np.sqrt(Delta)) * (Kk - F_psi)
        phi_vals[i] = phi
    
    # Ajustar φ para que sea monótona creciente
    # Las órbitas deben girar en sentido antihorario
    phi_vals = np.cumsum(np.abs(np.diff(np.concatenate([[0], phi_vals]))))
    
    # Convertir a coordenadas cartesianas
    x_vals = r_vals * np.cos(phi_vals)
    y_vals = r_vals * np.sin(phi_vals)
    
    return x_vals, y_vals, phi_vals, r_vals

# ============================================================================
# VERSIÓN ALTERNATIVA MÁS SIMPLE - USANDO ECUACIÓN DIFERENCIAL
# ============================================================================
def orbita_excent_imaginaria_simple(l, imag_e, M, num_points=10000):
    """
    Versión alternativa resolviendo la ecuación diferencial directamente
    """
    mu = M / l
    e = imag_e
    
    print(f"\nCalculando órbita simple con e = {e}i")
    
    # Resolver dφ/du = 1/sqrt(f(u))
    # donde f(u) = 2M(u - 1/(2M) + 2/l)[(u - 1/l)² + e²/l²]
    
    def f_u(u):
        term1 = u - (1/(2*M)) + (2/l)
        term2 = (u - 1/l)**2 + (e**2)/(l**2)
        return 2*M * term1 * term2
    
    # Rango de u (de grande a pequeño: desde cerca de la singularidad hasta infinito)
    u_vals = np.logspace(-2, 2, num_points)  # de 0.01 a 100
    
    # Filtrar valores donde f(u) > 0
    valid_mask = f_u(u_vals) > 0
    u_vals = u_vals[valid_mask]
    
    # Integrar φ(u) = ∫ (1/√f(u)) du
    integrand = 1.0 / np.sqrt(f_u(u_vals))
    
    # Usar integración acumulativa
    phi_vals = cumulative_trapezoid(integrand, u_vals, initial=0)
    
    # Ajustar para que φ aumente monótonamente
    r_vals = 1.0 / u_vals
    
    # Convertir a cartesianas
    x_vals = r_vals * np.cos(phi_vals)
    y_vals = r_vals * np.sin(phi_vals)
    
    return x_vals, y_vals, phi_vals, r_vals

# ============================================================================
# CREACIÓN DE LA FIGURA CON SÓLO MÉTODO SIMPLE (1 fila, 2 columnas)
# ============================================================================
fig, axes = plt.subplots(1, 2, figsize=(13, 6))  # Cambiado a 1 fila, 2 columnas

for idx, (imag_e, color, label) in enumerate(zip(imag_es, colors, labels)):
    print(f"\n{'='*80}")
    print(f"PROCESANDO: e = {imag_e}i")
    print(f"{'='*80}")
    
    # Método Simple (ecuación diferencial)
    try:
        ax = axes[idx]  # Ahora solo tenemos 1 fila
        x, y, phi, r = orbita_excent_imaginaria_simple(l, imag_e, M, 100000)
        ax.plot(x, y, color=color, linewidth=2.5, alpha=0.8, label=label)
        ax.set_title(f' e = {imag_e}i', fontsize=12)
        ax.grid(True, alpha=0.3)
        ax.axhline(0, color='k', alpha=0.2)
        ax.axvline(0, color='k', alpha=0.2)
        ax.set_aspect('equal')
        
        # Añadir horizonte y singularidad
        # Horizonte
        theta = np.linspace(0, 2*np.pi, 100)
        r_horizon = 2*M
        ax.plot(r_horizon*np.cos(theta), r_horizon*np.sin(theta), 
                'k--', alpha=0.6, linewidth=1.5, label='Horizonte')
        
        # Singularidad
        ax.plot(0, 0, 'ko', markersize=8, label='Singularidad')
        
        # Configurar límites
        max_range = 5
        ax.set_xlim(-max_range, max_range)
        ax.set_ylim(-max_range, max_range)
        ax.legend(loc='upper right', fontsize=8)
        ax.set_xlabel('x/M')
        ax.set_ylabel('y/M')
        
    except Exception as e:
        print(f"Error en método simple: {e}")

plt.suptitle('Geodésicas tipo-tiempo no ligadas con excentricidades imaginarias\n(E² > 1, l = 1, M = 0.3) ', 
             fontsize=16, fontweight='bold', y=1.02)
plt.tight_layout()
plt.show()
```
